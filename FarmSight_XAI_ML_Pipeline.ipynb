## Data Acquisition & Verification Stage
This section manages downloading the World Bank agricultural dataset, extracting the designated ICRISAT apportioned crop data, verifying its properties, and documenting data provenance.

```python
import requests
import hashlib
import zipfile
import os

# Configure URL and paths
zip_url = "https://reproducibility.worldbank.org/catalog/409/download/1152/RR_IND_2025_488.zip"
temp_zip_path = "/content/RR_IND_2025_488.zip"

# 1. & 2. Download the ZIP package with streaming and verify HTTP status
print(f"Downloading {zip_url} ...")
response = requests.get(zip_url, stream=True)
if response.status_code != 200:
    raise RuntimeError(f"Download failed with HTTP status code: {response.status_code}")

# 3. Save the downloaded ZIP temporarily under /content/
with open(temp_zip_path, 'wb') as f:
    for chunk in response.iter_content(chunk_size=1024 * 1024):
        if chunk:
            f.write(chunk)

# 4. Print size in MB and calculate SHA-256 checksum
file_size_mb = os.path.getsize(temp_zip_path) / (1024 * 1024)
print(f"Downloaded ZIP Size: {file_size_mb:.2f} MB")

sha256_hash = hashlib.sha256()
with open(temp_zip_path, 'rb') as f:
    for byte_block in iter(lambda: f.read(4096), b""):
        sha256_hash.update(byte_block)
zip_sha256 = sha256_hash.hexdigest()
print(f"ZIP SHA-256 Checksum: {zip_sha256}")
```

```python
# 5. & 6. Inspect ZIP contents without fully extracting and search for the target file
target_suffix = "Crops_Area Production Yield.csv"
target_path_sub = "raw/DLD/Apportioned/"
target_zip_path = None

with zipfile.ZipFile(temp_zip_path, 'r') as zip_ref:
    for file_info in zip_ref.infolist():
        if file_info.filename.endswith(target_suffix) and target_path_sub in file_info.filename:
            target_zip_path = file_info.filename
            break

if not target_zip_path:
    raise FileNotFoundError("Specified target apportioned crop CSV file was not found inside the ZIP package.")

print(f"Found target CSV inside ZIP: {target_zip_path}")

# 7. & 8. Extract only that CSV and save to raw data directory
destination_csv_path = "/content/drive/MyDrive/FarmSight_XAI/data/raw/icrisat_dld_crops_area_production_yield_apportioned.csv"

with zipfile.ZipFile(temp_zip_path, 'r') as zip_ref:
    with zip_ref.open(target_zip_path) as source_file:
        with open(destination_csv_path, 'wb') as dest_file:
            dest_file.write(source_file.read())

print(f"Extracted and saved to: {destination_csv_path}")
```

## Candidate Crop Coverage Analysis Stage
This section conducts a targeted read-only audit of spatial and temporal coverage for Rice, Maize, and Cotton to assist in scoping our experiments.

```python
import os
import pandas as pd

raw_path = "/content/drive/MyDrive/FarmSight_XAI/data/raw/icrisat_dld_crops_area_production_yield_apportioned.csv"
df_cov = pd.read_csv(raw_path)

candidates = ['RICE', 'MAIZE', 'COTTON']
crop_cols = {}

# 1. Identify the exact columns using case-insensitive match
for crop in candidates:
    crop_cols[crop] = {
        'area': [c for c in df_cov.columns if crop in c.upper() and 'AREA' in c.upper()][0],
        'production': [c for c in df_cov.columns if crop in c.upper() and 'PRODUCTION' in c.upper()][0],
        'yield': [c for c in df_cov.columns if crop in c.upper() and 'YIELD' in c.upper()][0]
    }

print("Matched Columns:")
for crop, cols in crop_cols.items():
    print(f"{crop}: {cols}")
```

```python
# 2. & 3. Calculate coverage and era-wise breakdowns
coverage_stats = []
era_bins = [
    (1966, 1989, "1966-1989"),
    (1990, 1999, "1990-1999"),
    (2000, 2009, "2000-2009"),
    (2010, 2017, "2010-2017")
]

for crop in candidates:
    y_col = crop_cols[crop]['yield']

    total_rows = len(df_cov)
    non_missing_yield = df_cov[y_col].notnull().sum()
    positive_yield = (df_cov[y_col] > 0).sum()
    zero_yield = (df_cov[y_col] == 0).sum()
    negative_yield = (df_cov[y_col] < 0).sum()

    # Filtering for positive yield subset
    pos_df = df_cov[df_cov[y_col] > 0]
    earliest_year = int(pos_df['Year'].min()) if not pos_df.empty else None
    latest_year = int(pos_df['Year'].max()) if not pos_df.empty else None
    unique_states = pos_df['State Name'].nunique() if not pos_df.empty else 0
    unique_districts = pos_df['Dist Name'].nunique() if not pos_df.empty else 0

    # Era breakdown for positive yield records
    era_counts = {}
    for start, end, label in era_bins:
        era_counts[label] = ((pos_df['Year'] >= start) & (pos_df['Year'] <= end)).sum()

    stat_dict = {
        'Crop': crop.capitalize(),
        'Total_Rows': total_rows,
        'Non_Missing_Yield': non_missing_yield,
        'Positive_Yield': positive_yield,
        'Zero_Yield': zero_yield,
        'Negative_Yield': negative_yield,
        'Earliest_Pos_Year': earliest_year,
        'Latest_Pos_Year': latest_year,
        'Unique_States_Pos': unique_states,
        'Unique_Districts_Pos': unique_districts
    }
    # Merge era counts
    for label, count in era_counts.items():
        stat_dict[f'Pos_Yield_Rows_{label}'] = count

    coverage_stats.append(stat_dict)

coverage_df = pd.DataFrame(coverage_stats)
print("\n--- Candidate Crop Coverage Summary ---")
display(coverage_df)
```

```python
import os
import hashlib
import pandas as pd
import numpy as np

raw_path = "/content/drive/MyDrive/FarmSight_XAI/data/raw/icrisat_dld_crops_area_production_yield_apportioned.csv"
interim_path = "/content/drive/MyDrive/FarmSight_XAI/data/interim/icrisat_rice_maize_cotton_long.csv"
conflicting_path = "/content/drive/MyDrive/FarmSight_XAI/results/tables/icrisat_conflicting_duplicates.csv"
report_path = "/content/drive/MyDrive/FarmSight_XAI/logs/icrisat_restructuring_report.txt"

# Expected raw SHA-256 to ensure read-only safety
expected_raw_hash = "7857decb1a196a54aac71a0f841af43c3bac98b3b90744aadca18b1859cea6f3"

def get_sha256(filepath):
    sha256 = hashlib.sha256()
    with open(filepath, 'rb') as f:
        for byte_block in iter(lambda: f.read(4096), b""):
            sha256.update(byte_block)
    return sha256.hexdigest()

# 1. Load the raw CSV
df_raw = pd.read_csv(raw_path)
raw_shape = df_raw.shape
print(f"Raw dataset loaded. Shape: {raw_shape}")
```

```python
# 2. & 3. Column identifiers and crop map
id_cols = {
    'Dist Code': 'district_code',
    'Dist Name': 'district_name',
    'State Code': 'state_code',
    'State Name': 'state_name',
    'Year': 'year'
}

crop_cols_map = {
    'Rice': {
        'area': 'RICE AREA (1000 ha)',
        'production': 'RICE PRODUCTION (1000 tons)',
        'yield': 'RICE YIELD (Kg per ha)'
    },
    'Maize': {
        'area': 'MAIZE AREA (1000 ha)',
        'production': 'MAIZE PRODUCTION (1000 tons)',
        'yield': 'MAIZE YIELD (Kg per ha)'
    },
    'Cotton': {
        'area': 'COTTON AREA (1000 ha)',
        'production': 'COTTON PRODUCTION (1000 tons)',
        'yield': 'COTTON YIELD (Kg per ha)'
    }
}

# 4. & 5. Reshape wide to long
long_records = []
for index, row in df_raw.iterrows():
    base_info = {id_cols[k]: row[k] for k in id_cols}
    for crop, metrics in crop_cols_map.items():
        rec = base_info.copy()
        rec['crop'] = crop
        rec['area_1000_ha'] = row[metrics['area']]
        rec['production_1000_tons'] = row[metrics['production']]
        rec['yield_kg_per_ha'] = row[metrics['yield']]
        long_records.append(rec)

df_long_all = pd.DataFrame(long_records)
```

```python
# 6. Filter & count removals (missing, zero, negative yield)
removal_stats = []
df_cleaned_list = []

for crop in ['Rice', 'Maize', 'Cotton']:
    crop_df = df_long_all[df_long_all['crop'] == crop].copy()

    total_before = len(crop_df)
    missing_mask = crop_df['yield_kg_per_ha'].isnull()
    zero_mask = crop_df['yield_kg_per_ha'] == 0
    negative_mask = crop_df['yield_kg_per_ha'] < 0

    missing_count = missing_mask.sum()
    zero_count = zero_mask.sum()
    negative_count = negative_mask.sum()

    # Keep positive non-missing
    keep_mask = crop_df['yield_kg_per_ha'] > 0
    cleaned_crop_df = crop_df[keep_mask].copy()

    df_cleaned_list.append(cleaned_crop_df)

    removal_stats.append({
        'Crop': crop,
        'Initial': total_before,
        'Removed_Missing': missing_count,
        'Removed_Zero': zero_count,
        'Removed_Negative': negative_count,
        'Retained': len(cleaned_crop_df)
    })

df_cleaned = pd.concat(df_cleaned_list, ignore_index=True)
removals_df = pd.DataFrame(removal_stats)
print("Removal Breakdown per Crop:")
display(removals_df)
```

```python
# 7. & 8. Check exact duplicate rows and duplicate keys
exact_dups = df_cleaned.duplicated().sum()
print(f"Number of exact duplicate rows: {exact_dups}")

# Remove identical duplicate rows
df_cleaned = df_cleaned.drop_duplicates().copy()

# Check for duplicate [district_code, year, crop] keys (conflict analysis)
key_cols = ['district_code', 'year', 'crop']
all_key_duplicates = df_cleaned[df_cleaned.duplicated(subset=key_cols, keep=False)]

# Separate completely conflicting keys
conflicting_records = pd.DataFrame(columns=df_cleaned.columns)
if not all_key_duplicates.empty:
    # Find rows where values differ but keys are identical
    grouped = all_key_duplicates.groupby(key_cols)
    conflicting_indices = []
    for keys, group in grouped:
        # If any of the variables differ within the grouped keys
        if group[['area_1000_ha', 'production_1000_tons', 'yield_kg_per_ha']].nunique().max() > 1:
            conflicting_indices.extend(group.index.tolist())

    if conflicting_indices:
        conflicting_records = df_cleaned.loc[conflicting_indices].copy()
        # Remove conflicting duplicates from our clean interim dataset
        df_cleaned = df_cleaned.drop(index=conflicting_indices).copy()

conflicting_count = len(conflicting_records)
print(f"Number of conflicting-key records isolated: {conflicting_count}")
```

```python
# 9, 10, 11 & 12. Yield Consistency calculation
# calculated_yield = production_1000_tons / area_1000_ha * 1000
# Checked only where area and production are positive (> 0)

calc_mask = (df_cleaned['area_1000_ha'] > 0) & (df_cleaned['production_1000_tons'] > 0)

df_cleaned['yield_consistency_flag'] = True
df_cleaned['percentage_difference'] = 0.0

if calc_mask.any():
    calculated_yield = (df_cleaned.loc[calc_mask, 'production_1000_tons'] / df_cleaned.loc[calc_mask, 'area_1000_ha']) * 1000
    reported_yield = df_cleaned.loc[calc_mask, 'yield_kg_per_ha']

    # Calculate percentage difference relative to reported yield (prevent divide by zero)
    pct_diff = np.where(
        reported_yield != 0,
        np.abs(reported_yield - calculated_yield) / reported_yield * 100,
        0.0
    )

    df_cleaned.loc[calc_mask, 'percentage_difference'] = pct_diff
    # Set flag to False if difference exceeds 10% tolerance limit
    df_cleaned.loc[calc_mask, 'yield_consistency_flag'] = pct_diff <= 10.0

flag_counts = df_cleaned['yield_consistency_flag'].value_counts()
print("Yield Consistency Flag Counts (True = within 10% tolerance, False = inconsistent):")
print(flag_counts)
```

```python
import os
import re
import time
import hashlib
import requests
import pandas as pd
from difflib import SequenceMatcher

input_path = "/content/drive/MyDrive/FarmSight_XAI/data/interim/icrisat_rice_maize_cotton_long.csv"
accepted_path = "/content/drive/MyDrive/FarmSight_XAI/data/interim/district_coordinates_accepted.csv"
review_path = "/content/drive/MyDrive/FarmSight_XAI/results/tables/district_geocoding_review.csv"
log_path = "/content/drive/MyDrive/FarmSight_XAI/logs/district_geocoding_report.txt"

def file_sha256(path):
    digest = hashlib.sha256()
    with open(path, "rb") as file:
        for block in iter(lambda: file.read(1024 * 1024), b""):
            digest.update(block)
    return digest.hexdigest()

def normalize(value):
    value = str(value).lower().strip()
    value = re.sub(r"\bdistrict\b", "", value)
    value = re.sub(r"[^a-z0-9]+", " ", value)
    return " ".join(value.split())

state_aliases = {
    "orissa": "odisha",
    "uttaranchal": "uttarakhand",
    "pondicherry": "puducherry",
}

def similarity(a, b):
    return SequenceMatcher(None, normalize(a), normalize(b)).ratio()

raw_hash_before = file_sha256(input_path)
data = pd.read_csv(input_path)

districts = (
    data[
        ["district_code", "district_name", "state_code", "state_name"]
    ]
    .drop_duplicates()
    .sort_values(["state_name", "district_name"])
    .reset_index(drop=True)
)

records = []
endpoint = "https://geocoding-api.open-meteo.com/v1/search"

for index, row in districts.iterrows():
    district = str(row["district_name"]).strip()
    state_original = str(row["state_name"]).strip()
    state_query = state_aliases.get(state_original.lower(), state_original)

    queries = [f"{district}, {state_query}", district]
    candidates = []
    error_message = ""

    for query in queries:
        try:
            response = requests.get(
                endpoint,
                params={
                    "name": query,
                    "count": 10,
                    "language": "en",
                    "countryCode": "IN",
                    "format": "json",
                },
                timeout=30,
            )
            response.raise_for_status()
            candidates = response.json().get("results", [])
            if candidates:
                break
        except Exception as error:
            error_message = str(error)

        time.sleep(0.15)

    scored = []
    for candidate in candidates:
        location_fields = [
            candidate.get("name", ""),
            candidate.get("admin2", ""),
            candidate.get("admin3", ""),
            candidate.get("admin4", ""),
        ]

        district_score = max(
            similarity(district, value)
            for value in location_fields
            if value
        )

        state_score = similarity(
            state_query,
            candidate.get("admin1", "")
        )

        combined_score = (0.80 * district_score) + (0.20 * state_score)
        scored.append((combined_score, district_score, state_score, candidate))

    if scored:
        scored.sort(key=lambda item: item[0], reverse=True)
        combined_score, district_score, state_score, best = scored[0]

        accepted = (
            district_score >= 0.85
            and state_score >= 0.60
            and best.get("latitude") is not None
            and best.get("longitude") is not None
        )

        record = {
            **row.to_dict(),
            "matched_name": best.get("name"),
            "matched_admin1": best.get("admin1"),
            "matched_admin2": best.get("admin2"),
            "matched_admin3": best.get("admin3"),
            "latitude": best.get("latitude"),
            "longitude": best.get("longitude"),
            "timezone": best.get("timezone"),
            "feature_code": best.get("feature_code"),
            "district_score": round(district_score, 4),
            "state_score": round(state_score, 4),
            "combined_score": round(combined_score, 4),
            "candidate_count": len(candidates),
            "match_status": "accepted" if accepted else "needs_review",
            "error": error_message,
        }
    else:
        record = {
            **row.to_dict(),
            "matched_name": None,
            "matched_admin1": None,
            "matched_admin2": None,
            "matched_admin3": None,
            "latitude": None,
            "longitude": None,
            "timezone": None,
            "feature_code": None,
            "district_score": 0,
            "state_score": 0,
            "combined_score": 0,
            "candidate_count": 0,
            "match_status": "unresolved",
            "error": error_message,
        }

    records.append(record)

    if (index + 1) % 25 == 0 or index + 1 == len(districts):
        print(f"Processed {index + 1}/{len(districts)} districts")

    time.sleep(0.15)

geocoded = pd.DataFrame(records)
accepted = geocoded[geocoded["match_status"] == "accepted"].copy()
review = geocoded[geocoded["match_status"] != "accepted"].copy()

accepted.to_csv(accepted_path, index=False)
review.to_csv(review_path, index=False)

raw_hash_after = file_sha256(input_path)
raw_unchanged = raw_hash_before == raw_hash_after

summary = f"""
District records attempted: {len(districts)}
Accepted automatically: {len(accepted)}
Needs review or unresolved: {len(review)}
Raw interim file unchanged: {raw_unchanged}
Accepted file: {accepted_path}
Review file: {review_path}

Coordinate meaning:
Open-Meteo geocoded representative locations are used.
They are not claimed to be official district centroids.
"""

with open(log_path, "w", encoding="utf-8") as file:
    file.write(summary.strip() + "\n")

outputs_exist = all(
    os.path.exists(path) and os.path.getsize(path) > 0
    for path in [accepted_path, review_path, log_path]
)

print(summary)
print("Output files created:", outputs_exist)

if raw_unchanged and outputs_exist and len(accepted) + len(review) == len(districts):
    print("DISTRICT GEOCODING STAGE VERIFIED")
else:
    print("DISTRICT GEOCODING STAGE FAILED")
```

```python
from pathlib import Path
import hashlib
import pandas as pd

base = Path("/content/drive/MyDrive/FarmSight_XAI")
long_path = base / "data/interim/icrisat_rice_maize_cotton_long.csv"
coords_path = base / "data/interim/district_coordinates_accepted.csv"

crop_output = base / "results/tables/geocoding_coverage_by_crop.csv"
state_output = base / "results/tables/geocoding_coverage_by_state.csv"
log_output = base / "logs/geocoding_coverage_report.txt"

def sha256_file(path):
    digest = hashlib.sha256()
    with open(path, "rb") as file:
        for block in iter(lambda: file.read(1024 * 1024), b""):
            digest.update(block)
    return digest.hexdigest()

hash_before = sha256_file(long_path)

agri = pd.read_csv(long_path)
coords = pd.read_csv(coords_path)

join_keys = ["state_code", "district_code"]

# Standardize ID types without modifying saved source files.
for column in join_keys:
    agri[column] = pd.to_numeric(agri[column], errors="raise").astype("Int64")
    coords[column] = pd.to_numeric(coords[column], errors="raise").astype("Int64")

duplicate_coordinate_keys = int(coords.duplicated(join_keys).sum())

coord_keys = coords[join_keys].drop_duplicates().copy()
coord_keys["coordinate_available"] = True

merged = agri.merge(
    coord_keys,
    on=join_keys,
    how="left",
    validate="many_to_one"
)

merged["coordinate_available"] = (
    merged["coordinate_available"].fillna(False).astype(bool)
)

def coverage_summary(group):
    total_rows = len(group)
    covered_rows = int(group["coordinate_available"].sum())

    total_districts = len(group[join_keys].drop_duplicates())
    covered_districts = len(
        group.loc[group["coordinate_available"], join_keys].drop_duplicates()
    )

    return {
        "total_rows": total_rows,
        "covered_rows": covered_rows,
        "row_coverage_pct": round(100 * covered_rows / total_rows, 2),
        "total_districts": total_districts,
        "covered_districts": covered_districts,
        "district_coverage_pct": round(
            100 * covered_districts / total_districts, 2
        ),
    }

crop_rows = []
for crop, group in merged.groupby("crop", dropna=False):
    crop_rows.append({"crop": crop, **coverage_summary(group)})

crop_coverage = pd.DataFrame(crop_rows).sort_values("crop")

state_rows = []
for (state_code, state_name), group in merged.groupby(
    ["state_code", "state_name"], dropna=False
):
    state_rows.append({
        "state_code": state_code,
        "state_name": state_name,
        **coverage_summary(group)
    })

state_coverage = (
    pd.DataFrame(state_rows)
    .sort_values(["district_coverage_pct", "state_name"])
)

overall = coverage_summary(merged)
states_below_70 = state_coverage[
    state_coverage["district_coverage_pct"] < 70
]

crop_output.parent.mkdir(parents=True, exist_ok=True)
log_output.parent.mkdir(parents=True, exist_ok=True)

crop_coverage.to_csv(crop_output, index=False)
state_coverage.to_csv(state_output, index=False)

hash_after = sha256_file(long_path)

report = f"""FarmSight XAI Geocoding Coverage Audit
Agricultural rows: {len(agri)}
Accepted coordinate rows: {len(coords)}
Duplicate coordinate keys: {duplicate_coordinate_keys}

Overall covered agricultural rows: {overall['covered_rows']}
Overall row coverage (%): {overall['row_coverage_pct']}
Overall covered districts: {overall['covered_districts']}
Overall districts: {overall['total_districts']}
Overall district coverage (%): {overall['district_coverage_pct']}

States below 70% district coverage: {len(states_below_70)}
Agricultural file hash unchanged: {hash_before == hash_after}

Coordinate interpretation:
Open-Meteo geocoded representative locations are used.
They are not claimed to be official district centroids.
"""

log_output.write_text(report, encoding="utf-8")

files_exist = all(
    path.exists() for path in [crop_output, state_output, log_output]
)

verified = (
    len(merged) == len(agri)
    and duplicate_coordinate_keys == 0
    and overall["covered_rows"] > 0
    and hash_before == hash_after
    and files_exist
)

print("Overall coverage:")
print(pd.Series(overall).to_string())

print("\nCoverage by crop:")
display(crop_coverage)

print("\nStates below 70% district coverage:")
display(states_below_70)

print("\nOutput files created:", files_exist)
print("Agricultural file hash unchanged:", hash_before == hash_after)
print("Duplicate coordinate keys:", duplicate_coordinate_keys)

if verified:
    print("\nGEOCODING COVERAGE AUDIT VERIFIED")
else:
    print("\nGEOCODING COVERAGE AUDIT FAILED")
```

```python
from pathlib import Path
import gzip
import hashlib
import json
import time
import requests
import pandas as pd

base = Path("/content/drive/MyDrive/FarmSight_XAI")

coords_path = base / "data/interim/district_coordinates_accepted.csv"
agri_path = base / "data/interim/icrisat_rice_maize_cotton_long.csv"

raw_weather_dir = base / "data/raw/era5_daily_by_district"
district_feature_dir = base / "data/interim/era5_annual_by_district"

combined_output = (
    base / "data/interim/era5_district_year_weather_features.csv"
)
manifest_output = base / "logs/weather_acquisition_manifest.csv"
report_output = base / "logs/weather_full_acquisition_report.txt"

raw_weather_dir.mkdir(parents=True, exist_ok=True)
district_feature_dir.mkdir(parents=True, exist_ok=True)

def sha256_file(path):
    digest = hashlib.sha256()
    with open(path, "rb") as file:
        for block in iter(lambda: file.read(1024 * 1024), b""):
            digest.update(block)
    return digest.hexdigest()

coords_hash_before = sha256_file(coords_path)
agri_hash_before = sha256_file(agri_path)

coords = pd.read_csv(coords_path)
agri = pd.read_csv(agri_path)

lat_col = next(
    c for c in coords.columns
    if "latitude" in c.lower() or c.lower() == "lat"
)
lon_col = next(
    c for c in coords.columns
    if "longitude" in c.lower() or c.lower() in {"lon", "lng"}
)

start_year = int(agri["year"].min())
end_year = int(agri["year"].max())
expected_years = end_year - start_year + 1

session = requests.Session()
manifest_rows = []

def create_annual_features(payload, row):
    daily = pd.DataFrame(payload["daily"])

    daily["date"] = pd.to_datetime(daily["time"])
    daily["year"] = daily["date"].dt.year
    daily["month"] = daily["date"].dt.month

    daily["hot_day"] = daily["temperature_2m_max"] >= 35
    daily["dry_day"] = daily["precipitation_sum"] < 1
    daily["heavy_rain_day"] = daily["precipitation_sum"] >= 50

    annual = daily.groupby("year").agg(
        annual_mean_temp_c=("temperature_2m_mean", "mean"),
        annual_max_temp_c=("temperature_2m_max", "max"),
        annual_min_temp_c=("temperature_2m_min", "min"),
        annual_precipitation_mm=("precipitation_sum", "sum"),
        annual_et0_mm=("et0_fao_evapotranspiration", "sum"),
        hot_days=("hot_day", "sum"),
        dry_days=("dry_day", "sum"),
        heavy_rain_days=("heavy_rain_day", "sum")
    ).reset_index()

    monsoon = (
        daily[daily["month"].between(6, 9)]
        .groupby("year")
        .agg(
            monsoon_mean_temp_c=("temperature_2m_mean", "mean"),
            monsoon_precipitation_mm=("precipitation_sum", "sum"),
            monsoon_et0_mm=("et0_fao_evapotranspiration", "sum")
        )
        .reset_index()
    )

    annual = annual.merge(monsoon, on="year", how="left")

    annual["annual_water_balance_mm"] = (
        annual["annual_precipitation_mm"] -
        annual["annual_et0_mm"]
    )

    annual["monsoon_water_balance_mm"] = (
        annual["monsoon_precipitation_mm"] -
        annual["monsoon_et0_mm"]
    )

    annual.insert(0, "district_name", row["district_name"])
    annual.insert(0, "district_code", int(row["district_code"]))
    annual.insert(0, "state_name", row["state_name"])
    annual.insert(0, "state_code", int(row["state_code"]))

    annual["latitude"] = float(row[lat_col])
    annual["longitude"] = float(row[lon_col])

    return annual

for position, (_, row) in enumerate(coords.iterrows(), start=1):
    state_code = int(row["state_code"])
    district_code = int(row["district_code"])

    file_id = f"state_{state_code}_district_{district_code}"

    raw_path = raw_weather_dir / f"{file_id}.json.gz"
    annual_path = district_feature_dir / f"{file_id}.csv"

    # Resume safely if this district has already been verified.
    if annual_path.exists():
        existing = pd.read_csv(annual_path)

        if (
            len(existing) == expected_years
            and existing["year"].nunique() == expected_years
        ):
            manifest_rows.append({
                "state_code": state_code,
                "district_code": district_code,
                "district_name": row["district_name"],
                "status": "existing_verified",
                "attempts": 0,
                "elapsed_seconds": 0,
                "error": ""
            })

            print(
                f"[{position}/{len(coords)}] "
                f"{row['district_name']}: already verified"
            )
            continue

    params = {
        "latitude": float(row[lat_col]),
        "longitude": float(row[lon_col]),
        "start_date": f"{start_year}-01-01",
        "end_date": f"{end_year}-12-31",
        "daily": [
            "temperature_2m_mean",
            "temperature_2m_max",
            "temperature_2m_min",
            "precipitation_sum",
            "et0_fao_evapotranspiration"
        ],
        "timezone": "Asia/Kolkata",
        "models": "era5"
    }

    success = False
    last_error = ""
    request_start = time.time()

    for attempt in range(1, 5):
        try:
            response = session.get(
                "https://archive-api.open-meteo.com/v1/archive",
                params=params,
                timeout=300
            )
            response.raise_for_status()

            payload = response.json()
            annual = create_annual_features(payload, row)

            feature_columns = [
                c for c in annual.columns
                if c not in {
                    "state_code", "state_name",
                    "district_code", "district_name",
                    "year", "latitude", "longitude"
                }
            ]

            if len(annual) != expected_years:
                raise ValueError(
                    f"Expected {expected_years} years, got {len(annual)}"
                )

            if annual[feature_columns].isna().any().any():
                raise ValueError("Missing annual weather features detected")

            with gzip.open(raw_path, "wt", encoding="utf-8") as file:
                json.dump(payload, file)

            annual.to_csv(annual_path, index=False)

            success = True
            break

        except Exception as error:
            last_error = str(error)
            wait_seconds = 5 * (2 ** (attempt - 1))

            print(
                f"[{position}/{len(coords)}] "
                f"{row['district_name']} attempt {attempt} failed: "
                f"{last_error}"
            )

            if attempt < 4:
                time.sleep(wait_seconds)

    elapsed = round(time.time() - request_start, 2)

    manifest_rows.append({
        "state_code": state_code,
        "district_code": district_code,
        "district_name": row["district_name"],
        "status": "downloaded_verified" if success else "failed",
        "attempts": attempt,
        "elapsed_seconds": elapsed,
        "error": "" if success else last_error
    })

    pd.DataFrame(manifest_rows).to_csv(
        manifest_output, index=False
    )

    print(
        f"[{position}/{len(coords)}] {row['district_name']}: "
        f"{'VERIFIED' if success else 'FAILED'}"
    )

    # Polite delay for the public API.
    time.sleep(1)

# Combine every verified district file.
annual_files = sorted(district_feature_dir.glob("*.csv"))
combined = pd.concat(
    [pd.read_csv(path) for path in annual_files],
    ignore_index=True
)

combined.to_csv(combined_output, index=False)

manifest = pd.DataFrame(manifest_rows)
failed_count = int((manifest["status"] == "failed").sum())

duplicate_keys = int(
    combined.duplicated(
        ["state_code", "district_code", "year"]
    ).sum()
)

missing_values = int(combined.isna().sum().sum())
expected_rows = len(coords) * expected_years

source_hashes_unchanged = (
    sha256_file(coords_path) == coords_hash_before
    and sha256_file(agri_path) == agri_hash_before
)

verified = (
    failed_count == 0
    and len(annual_files) == len(coords)
    and len(combined) == expected_rows
    and duplicate_keys == 0
    and missing_values == 0
    and source_hashes_unchanged
)

report = f"""Full ERA5 Weather Acquisition
Districts expected: {len(coords)}
District files created: {len(annual_files)}
Period: {start_year}-{end_year}
Years per district: {expected_years}
Expected combined rows: {expected_rows}
Actual combined rows: {len(combined)}
Failed districts: {failed_count}
Duplicate district-year keys: {duplicate_keys}
Missing values: {missing_values}
Source hashes unchanged: {source_hashes_unchanged}
Acquisition verified: {verified}
"""

report_output.write_text(report, encoding="utf-8")

print("\n" + report)

if verified:
    print("FULL ERA5 WEATHER ACQUISITION VERIFIED")
else:
    print("WEATHER ACQUISITION INCOMPLETE — RERUN THIS CELL TO RESUME")
```

```python
import os
import time
import zipfile
import hashlib
from pathlib import Path
import pandas as pd
import xarray as xr
from google.colab import userdata
import cdsapi

base = Path("/content/drive/MyDrive/FarmSight_XAI")
coords_path = base / "data/interim/district_coordinates_accepted.csv"
output_dir = base / "data/raw/cds_fullspan_pilot"
output_dir.mkdir(parents=True, exist_ok=True)

log_path = base / "logs/cds_fullspan_pilot_validation.txt"

# 1. Load coordinates and select Chittoor (first district)
coords = pd.read_csv(coords_path)
sample = coords.iloc[0]

# Configure Target Path
target_file = output_dir / "cds_era5_fullspan_pilot_1966_2017.nc"
extract_dir = output_dir / "extracted"
extract_dir.mkdir(parents=True, exist_ok=True)

start_year = 1966
end_year = 2017

request = {
    "variable": ["2m_temperature", "total_precipitation"],
    "location": {
        "longitude": float(sample["longitude"]),
        "latitude": float(sample["latitude"])
    },
    "date": [f"{start_year}-01-01/{end_year}-12-31"],
    "data_format": "netcdf"
}

# 2. Check if already exists to avoid redundant downloads
download_needed = True
elapsed_time = 0.0

if target_file.is_file() and target_file.stat().st_size > 0:
    print(f"Existing download found at {target_file}. Skipping API request.")
    download_needed = False
else:
    print(f"Requesting full-span data for {sample['district_name']} ({start_year}-{end_year})...", flush=True)
    cds_client = cdsapi.Client(
        url="https://cds.climate.copernicus.eu/api",
        key=userdata.get("CDS_API_KEY").strip(),
        debug=False
    )
    start_time = time.time()
    cds_client.retrieve(
        "reanalysis-era5-single-levels-timeseries",
        request,
        str(target_file)
    )
    elapsed_time = round(time.time() - start_time, 2)
    print(f"Download completed in {elapsed_time} seconds.")
```

```python
# 4. Validation of Variables & Temporal Completeness
expected_index = pd.date_range(
    start=f"{start_year}-01-01 00:00:00",
    end=f"{end_year}-12-31 23:00:00",
    freq="h",
    tz="UTC"
)

time_dim = 'valid_time' if 'valid_time' in ds_full.dims else ('time' if 'time' in ds_full.dims else None)
assert time_dim is not None, "Time dimension not found in dataset."

actual_times = pd.DatetimeIndex(ds_full[time_dim].values)
if actual_times.tz is None:
    actual_times = actual_times.tz_localize("UTC")

timestamps_match = actual_times.equals(expected_index)
expected_hours = len(expected_index)
actual_hours = len(actual_times)

t2m_missing = int(ds_full["t2m"].isnull().sum().values)
tp_missing = int(ds_full["tp"].isnull().sum().values)
t2m_units = ds_full["t2m"].attrs.get("units", "N/A")
tp_units = ds_full["tp"].attrs.get("units", "N/A")

checks_passed = (
    timestamps_match
    and t2m_missing == 0
    and tp_missing == 0
)

# Determine coordinates
grid_lat = float(ds_full["latitude"].values) if "latitude" in ds_full.coords else float(sample["latitude"])
grid_lon = float(ds_full["longitude"].values) if "longitude" in ds_full.coords else float(sample["longitude"])

report_text = f"""CDS Full-Span Feasibility Pilot Report
======================================
District: {sample['district_name']}
State: {sample['state_name']}
Target Coordinates (Query): Lat {sample['latitude']}, Lon {sample['longitude']}
Grid Coordinates (Returned): Lat {grid_lat}, Lon {grid_lon}
File Name: {target_file.name}
File Size: {target_file.stat().st_size} bytes
Format (Is ZIP): {is_zip}

Temporal Audit:
---------------
Target Period: {start_year}-01-01 00:00 UTC to {end_year}-12-31 23:00 UTC
Expected Hours: {expected_hours}
Actual Hours: {actual_hours}
Timestamps Match Exactly: {timestamps_match}

Variable Quality Controls:
--------------------------
2m Temperature ('t2m'):
  - Units: {t2m_units}
  - Missing Values: {t2m_missing}
Total Precipitation ('tp'):
  - Units: {tp_units}
  - Missing Values: {tp_missing}

Validation Status: {'SUCCESS' if checks_passed else 'FAILED'}
"""

with open(log_path, "w", encoding="utf-8") as f:
    f.write(report_text)

print(report_text)

if checks_passed:
    print("CDS FULL-SPAN PILOT VERIFIED")
else:
    print("CDS FULL-SPAN PILOT VALIDATION FAILED")
```

```python
from pathlib import Path
import time
import requests
import xarray as xr

base = Path("/content/drive/MyDrive/FarmSight_XAI")
output_dir = base / "data/raw/nasa_power_india/T2M_tiles"
output_dir.mkdir(parents=True, exist_ok=True)

url = "https://power.larc.nasa.gov/api/temporal/monthly/regional"

latitude_tiles = [(6, 16), (16, 26), (26, 36), (36, 38)]
longitude_tiles = [(68, 78), (78, 88), (88, 98)]

verified_files = []

for lat_min, lat_max in latitude_tiles:
    for lon_min, lon_max in longitude_tiles:

        output_file = output_dir / (
            f"T2M_lat_{lat_min}_{lat_max}_lon_{lon_min}_{lon_max}_"
            f"1981_2017.nc"
        )

        # Resume safely if a valid tile already exists
        if output_file.exists():
            try:
                existing = xr.open_dataset(output_file)
                if "T2M" in existing.data_vars:
                    existing.close()
                    verified_files.append(output_file)
                    print("Already verified:", output_file.name)
                    continue
                existing.close()
            except Exception:
                pass

        params = {
            "parameters": "T2M",
            "community": "AG",
            "latitude-min": lat_min,
            "latitude-max": lat_max,
            "longitude-min": lon_min,
            "longitude-max": lon_max,
            "start": 1981,
            "end": 2017,
            "format": "NETCDF",
        }

        print("\nRequesting:", output_file.name)

        success = False

        for attempt in range(1, 4):
            response = requests.get(url, params=params, timeout=600)
            print(
                f"Attempt {attempt} | HTTP {response.status_code} | "
                f"{len(response.content)} bytes"
            )

            if response.status_code == 200:
                output_file.write_bytes(response.content)

                try:
                    dataset = xr.open_dataset(output_file)

                    if "T2M" not in dataset.data_vars:
                        raise ValueError(
                            f"T2M missing. Variables: {list(dataset.data_vars)}"
                        )

                    missing = int(dataset["T2M"].isnull().sum().values)

                    print("Dimensions:", dict(dataset.sizes))
                    print("Missing T2M values:", missing)

                    dataset.close()

                    verified_files.append(output_file)
                    print("VERIFIED:", output_file.name)

                    success = True
                    break

                except Exception as error:
                    print("Downloaded-file validation failed:", error)
                    if output_file.exists():
                        output_file.unlink()

            elif response.status_code == 429:
                print("Rate limited. Waiting before retry...")
                time.sleep(30 * attempt)

            else:
                print(response.text[:1000])
                break

        if not success:
            raise RuntimeError(
                f"Tile failed: lat {lat_min}-{lat_max}, "
                f"lon {lon_min}-{lon_max}"
            )

        time.sleep(3)

print("\nT2M TILE DOWNLOAD SUMMARY")
print("Expected tiles: 12")
print("Verified tiles:", len(verified_files))

if len(verified_files) != 12:
    raise RuntimeError("Not all 12 T2M tiles were verified.")

print("NASA POWER T2M INDIA TILES VERIFIED")
```

```python
from pathlib import Path
import time
import requests
import xarray as xr

base = Path("/content/drive/MyDrive/FarmSight_XAI")
root = base / "data/raw/nasa_power_india"

url = "https://power.larc.nasa.gov/api/temporal/monthly/regional"

# Maximum temperature, minimum temperature, precipitation
parameters = ["T2M_MAX", "T2M_MIN", "PRECTOTCORR"]

latitude_tiles = [(6, 16), (16, 26), (26, 36), (36, 38)]
longitude_tiles = [(68, 78), (78, 88), (88, 98)]

summary = {}

for parameter in parameters:

    output_dir = root / f"{parameter}_tiles"
    output_dir.mkdir(parents=True, exist_ok=True)

    verified = 0
    total_missing = 0

    print(f"\n{'=' * 60}")
    print("DOWNLOADING:", parameter)
    print("=" * 60)

    for lat_min, lat_max in latitude_tiles:
        for lon_min, lon_max in longitude_tiles:

            output_file = output_dir / (
                f"{parameter}_lat_{lat_min}_{lat_max}_"
                f"lon_{lon_min}_{lon_max}_1981_2017.nc"
            )

            # Resume from previously verified files
            if output_file.exists():
                try:
                    dataset = xr.open_dataset(output_file)

                    if parameter in dataset.data_vars:
                        missing = int(
                            dataset[parameter].isnull().sum().values
                        )
                        dataset.close()

                        if missing == 0:
                            verified += 1
                            print("Already verified:", output_file.name)
                            continue
                    else:
                        dataset.close()

                except Exception:
                    pass

            request_params = {
                "parameters": parameter,
                "community": "AG",
                "latitude-min": lat_min,
                "latitude-max": lat_max,
                "longitude-min": lon_min,
                "longitude-max": lon_max,
                "start": 1981,
                "end": 2017,
                "format": "NETCDF",
            }

            print("\nRequesting:", output_file.name)
            success = False

            for attempt in range(1, 4):

                response = requests.get(
                    url,
                    params=request_params,
                    timeout=600
                )

                print(
                    f"Attempt {attempt} | "
                    f"HTTP {response.status_code} | "
                    f"{len(response.content)} bytes"
                )

                if response.status_code == 200:

                    output_file.write_bytes(response.content)

                    try:
                        dataset = xr.open_dataset(output_file)

                        if parameter not in dataset.data_vars:
                            raise ValueError(
                                f"{parameter} missing. "
                                f"Variables: {list(dataset.data_vars)}"
                            )

                        missing = int(
                            dataset[parameter].isnull().sum().values
                        )

                        print("Dimensions:", dict(dataset.sizes))
                        print("Missing values:", missing)

                        dataset.close()

                        if missing != 0:
                            raise ValueError(
                                f"{parameter} contains missing values."
                            )

                        verified += 1
                        success = True

                        print("VERIFIED:", output_file.name)
                        break

                    except Exception as error:
                        print("Validation failed:", error)

                        if output_file.exists():
                            output_file.unlink()

                elif response.status_code == 429:
                    print("Rate limited. Waiting before retry...")
                    time.sleep(30 * attempt)

                else:
                    print(response.text[:1000])
                    break

            if not success:
                raise RuntimeError(
                    f"{parameter} tile failed: "
                    f"lat {lat_min}-{lat_max}, "
                    f"lon {lon_min}-{lon_max}"
                )

            time.sleep(3)

    # Final parameter verification
    files = sorted(output_dir.glob("*.nc"))

    verified_again = 0
    total_missing = 0

    for file in files:
        dataset = xr.open_dataset(file)

        if parameter in dataset.data_vars:
            missing = int(
                dataset[parameter].isnull().sum().values
            )
            total_missing += missing

            if missing == 0:
                verified_again += 1

        dataset.close()

    summary[parameter] = {
        "files": len(files),
        "verified": verified_again,
        "missing": total_missing,
    }

print("\nFINAL NASA POWER DOWNLOAD SUMMARY")
print("=" * 60)

all_verified = True

for parameter, result in summary.items():
    print(
        f"{parameter}: "
        f"files={result['files']}, "
        f"verified={result['verified']}, "
        f"missing={result['missing']}"
    )

    if (
        result["files"] != 12
        or result["verified"] != 12
        or result["missing"] != 0
    ):
        all_verified = False

if all_verified:
    print("ALL NASA POWER WEATHER VARIABLES VERIFIED")
else:
    print("NASA POWER WEATHER DATA NEEDS CORRECTION")
```

```python
from pathlib import Path
import hashlib
import pandas as pd

base = Path("/content/drive/MyDrive/FarmSight_XAI")

crop_path = (
    base / "data/interim/"
    "icrisat_rice_maize_cotton_long.csv"
)

weather_path = (
    base / "data/interim/"
    "nasa_power_district_annual_1981_2017.csv"
)

output_path = (
    base / "data/processed/"
    "farmsight_crop_weather_1981_2017.csv"
)

coverage_path = (
    base / "results/tables/"
    "crop_weather_join_coverage.csv"
)

report_path = (
    base / "logs/"
    "crop_weather_join_report.txt"
)

for path in [output_path, coverage_path, report_path]:
    path.parent.mkdir(parents=True, exist_ok=True)

# ---------------------------------------------------------
# 1. Load datasets
# ---------------------------------------------------------

crops = pd.read_csv(crop_path)
weather = pd.read_csv(weather_path)

print("Crop dataset shape:", crops.shape)
print("Weather dataset shape:", weather.shape)

# ---------------------------------------------------------
# 2. Validate required columns
# ---------------------------------------------------------

join_keys = [
    "state_code",
    "district_code",
    "year",
]

crop_required = join_keys + [
    "state_name",
    "district_name",
    "crop",
    "area_1000_ha",
    "production_1000_tons",
    "yield_kg_per_ha",
    "yield_consistency_flag",
]

weather_features = [
    "district_latitude",
    "district_longitude",
    "nasa_grid_latitude",
    "nasa_grid_longitude",
    "grid_distance_km",
    "annual_temp_mean_c",
    "annual_temp_max_c",
    "annual_temp_min_c",
    "annual_temp_range_c",
    "annual_precip_mm",
    "monsoon_precip_mm",
    "monsoon_temp_mean_c",
    "monthly_precip_rate_std_mm_day",
]

weather_required = join_keys + weather_features

missing_crop_columns = [
    column for column in crop_required
    if column not in crops.columns
]

missing_weather_columns = [
    column for column in weather_required
    if column not in weather.columns
]

if missing_crop_columns:
    raise ValueError(
        f"Missing crop columns: {missing_crop_columns}"
    )

if missing_weather_columns:
    raise ValueError(
        f"Missing weather columns: "
        f"{missing_weather_columns}"
    )

# ---------------------------------------------------------
# 3. Standardize key types and restrict common period
# ---------------------------------------------------------

for frame in [crops, weather]:
    frame["state_code"] = frame["state_code"].astype(int)
    frame["district_code"] = frame["district_code"].astype(int)
    frame["year"] = frame["year"].astype(int)

crop_period = crops[
    crops["year"].between(1981, 2017)
].copy()

weather_for_join = weather[
    weather_required
].copy()

weather_duplicate_keys = int(
    weather_for_join.duplicated(join_keys).sum()
)

if weather_duplicate_keys != 0:
    raise ValueError(
        f"Weather contains {weather_duplicate_keys} "
        f"duplicate join keys."
    )

# ---------------------------------------------------------
# 4. Perform audited many-to-one join
# ---------------------------------------------------------

joined_audit = crop_period.merge(
    weather_for_join,
    on=join_keys,
    how="left",
    validate="many_to_one",
    indicator=True,
)

matched = joined_audit[
    joined_audit["_merge"] == "both"
].copy()

unmatched = joined_audit[
    joined_audit["_merge"] == "left_only"
].copy()

matched = matched.drop(columns="_merge")

# ---------------------------------------------------------
# 5. Validate joined dataset
# ---------------------------------------------------------

model_key = [
    "state_code",
    "district_code",
    "year",
    "crop",
]

duplicate_model_keys = int(
    matched.duplicated(model_key).sum()
)

missing_weather_values = int(
    matched[weather_features]
    .isna()
    .sum()
    .sum()
)

matched_districts = (
    matched[
        ["state_code", "district_code"]
    ]
    .drop_duplicates()
    .shape[0]
)

matched_year_min = int(matched["year"].min())
matched_year_max = int(matched["year"].max())

join_coverage_pct = (
    len(matched) / len(crop_period) * 100
    if len(crop_period) > 0
    else 0
)

# Treat the consistency flag robustly
consistency_as_text = (
    matched["yield_consistency_flag"]
    .astype(str)
    .str.strip()
    .str.lower()
)

consistent_rows = int(
    consistency_as_text.eq("true").sum()
)

inconsistent_rows = int(
    consistency_as_text.eq("false").sum()
)

unknown_consistency_rows = int(
    len(matched)
    - consistent_rows
    - inconsistent_rows
)

# ---------------------------------------------------------
# 6. Coverage table by crop
# ---------------------------------------------------------

coverage_records = []

for crop_name, crop_group in crop_period.groupby("crop"):

    matched_crop = matched[
        matched["crop"] == crop_name
    ]

    coverage_records.append({
        "crop": crop_name,
        "eligible_crop_rows_1981_2017":
            len(crop_group),
        "matched_rows":
            len(matched_crop),
        "unmatched_rows":
            len(crop_group) - len(matched_crop),
        "row_coverage_pct":
            round(
                len(matched_crop)
                / len(crop_group)
                * 100,
                3,
            ),
        "matched_districts":
            matched_crop[
                ["state_code", "district_code"]
            ]
            .drop_duplicates()
            .shape[0],
        "year_min":
            int(matched_crop["year"].min())
            if len(matched_crop) else None,
        "year_max":
            int(matched_crop["year"].max())
            if len(matched_crop) else None,
    })

coverage = pd.DataFrame(coverage_records)

# ---------------------------------------------------------
# 7. Save outputs
# ---------------------------------------------------------

matched.to_csv(
    output_path,
    index=False
)

coverage.to_csv(
    coverage_path,
    index=False
)

def sha256_file(path):
    digest = hashlib.sha256()

    with open(path, "rb") as file:
        for block in iter(
            lambda: file.read(1024 * 1024),
            b"",
        ):
            digest.update(block)

    return digest.hexdigest()

crop_hash = sha256_file(crop_path)
weather_hash = sha256_file(weather_path)
joined_hash = sha256_file(output_path)

crop_period_rows = len(crop_period)
matched_rows = len(matched)
unmatched_rows = len(unmatched)

report = f"""FarmSight Crop-Weather Join Report
=================================
Join keys: state_code, district_code, year
Join validation: many crop rows to one weather row
Common period: 1981-2017

Crop rows in common period: {crop_period_rows}
Matched crop-weather rows: {matched_rows}
Unmatched crop rows: {unmatched_rows}
Join row coverage: {join_coverage_pct:.3f}%

Matched districts: {matched_districts}
Matched year range: {matched_year_min}-{matched_year_max}

Duplicate district-year-crop keys: {duplicate_model_keys}
Missing weather feature values: {missing_weather_values}

Yield-consistency flag:
Consistent rows: {consistent_rows}
Inconsistent rows: {inconsistent_rows}
Unknown flag rows: {unknown_consistency_rows}

Input crop SHA-256:
{crop_hash}

Input weather SHA-256:
{weather_hash}

Joined output SHA-256:
{joined_hash}
"""

report_path.write_text(
    report,
    encoding="utf-8"
)

# ---------------------------------------------------------
# 8. Final verification
# ---------------------------------------------------------

print("\n" + report)

print("\nCoverage by crop:")
print(coverage.to_string(index=False))

print("\nSaved joined dataset:", output_path)
print("Saved coverage table:", coverage_path)
print("Saved join report:", report_path)

checks_passed = (
    matched_rows > 0
    and matched_districts > 0
    and matched_year_min == 1981
    and matched_year_max == 2017
    and duplicate_model_keys == 0
    and missing_weather_values == 0
    and unknown_consistency_rows == 0
)

if checks_passed:
    print("CROP-WEATHER JOIN VERIFIED")
else:
    print("CROP-WEATHER JOIN NEEDS CORRECTION")
```

```python
from pathlib import Path
import hashlib
import numpy as np
import pandas as pd

base = Path("/content/drive/MyDrive/FarmSight_XAI")

input_path = (
    base / "data/processed/"
    "farmsight_crop_weather_1981_2017.csv"
)

output_path = (
    base / "data/processed/"
    "farmsight_modeling_dataset_labeled.csv"
)

distribution_path = (
    base / "results/tables/"
    "crop_stress_label_distribution.csv"
)

report_path = (
    base / "logs/"
    "crop_stress_target_definition_report.txt"
)

for path in [
    output_path,
    distribution_path,
    report_path,
]:
    path.parent.mkdir(parents=True, exist_ok=True)

data = pd.read_csv(input_path)

print("Joined input shape:", data.shape)

# ---------------------------------------------------------
# 1. Validate required columns
# ---------------------------------------------------------

required_columns = [
    "state_code",
    "state_name",
    "district_code",
    "district_name",
    "year",
    "crop",
    "yield_kg_per_ha",
    "yield_consistency_flag",
    "annual_temp_mean_c",
    "annual_temp_max_c",
    "annual_temp_min_c",
    "annual_temp_range_c",
    "annual_precip_mm",
    "monsoon_precip_mm",
    "monsoon_temp_mean_c",
    "monthly_precip_rate_std_mm_day",
]

missing_columns = [
    column for column in required_columns
    if column not in data.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

# ---------------------------------------------------------
# 2. Exclude yield-inconsistent records
# ---------------------------------------------------------

consistency_flag = (
    data["yield_consistency_flag"]
    .astype(str)
    .str.strip()
    .str.lower()
)

consistent_data = data[
    consistency_flag.eq("true")
].copy()

excluded_inconsistent = (
    len(data) - len(consistent_data)
)

if (
    consistent_data["yield_kg_per_ha"].isna().any()
    or (consistent_data["yield_kg_per_ha"] <= 0).any()
):
    raise ValueError(
        "Consistent data still contains invalid yield values."
    )

# ---------------------------------------------------------
# 3. Calculate causal five-observation baseline
# ---------------------------------------------------------

group_columns = [
    "state_code",
    "district_code",
    "crop",
]

consistent_data = consistent_data.sort_values(
    group_columns + ["year"]
).reset_index(drop=True)

consistent_data["yield_baseline_previous_5"] = (
    consistent_data
    .groupby(
        group_columns,
        sort=False,
    )["yield_kg_per_ha"]
    .transform(
        lambda series:
            series
            .shift(1)
            .rolling(
                window=5,
                min_periods=5,
            )
            .median()
    )
)

baseline_available = (
    consistent_data[
        "yield_baseline_previous_5"
    ].notna()
)

labeled = consistent_data[
    baseline_available
].copy()

excluded_no_baseline = (
    len(consistent_data) - len(labeled)
)

# ---------------------------------------------------------
# 4. Calculate yield anomaly and stress label
# ---------------------------------------------------------

labeled["yield_anomaly_pct"] = (
    (
        labeled["yield_kg_per_ha"]
        - labeled["yield_baseline_previous_5"]
    )
    / labeled["yield_baseline_previous_5"]
    * 100
)

labeled["crop_stress_label"] = np.select(
    [
        labeled["yield_anomaly_pct"] <= -20,
        (
            labeled["yield_anomaly_pct"] > -20
        )
        & (
            labeled["yield_anomaly_pct"] <= -10
        ),
    ],
    [
        "High",
        "Moderate",
    ],
    default="Low",
)

label_order = [
    "Low",
    "Moderate",
    "High",
]

labeled["crop_stress_label"] = pd.Categorical(
    labeled["crop_stress_label"],
    categories=label_order,
    ordered=True,
)

# ---------------------------------------------------------
# 5. Validate target dataset
# ---------------------------------------------------------

model_key = [
    "state_code",
    "district_code",
    "year",
    "crop",
]

duplicate_keys = int(
    labeled.duplicated(model_key).sum()
)

target_missing = int(
    labeled["crop_stress_label"].isna().sum()
)

baseline_missing = int(
    labeled[
        "yield_baseline_previous_5"
    ].isna().sum()
)

infinite_anomalies = int(
    np.isinf(
        labeled["yield_anomaly_pct"]
    ).sum()
)

future_baseline_avoided = True

# ---------------------------------------------------------
# 6. Create label-distribution tables
# ---------------------------------------------------------

overall_distribution = (
    labeled["crop_stress_label"]
    .value_counts(sort=False)
    .reindex(label_order, fill_value=0)
    .rename_axis("crop_stress_label")
    .reset_index(name="count")
)

overall_distribution["percentage"] = (
    overall_distribution["count"]
    / len(labeled)
    * 100
).round(3)

distribution_by_crop = (
    labeled
    .groupby(
        ["crop", "crop_stress_label"],
        observed=False,
    )
    .size()
    .reset_index(name="count")
)

crop_totals = (
    distribution_by_crop
    .groupby("crop")["count"]
    .transform("sum")
)

distribution_by_crop["percentage"] = (
    distribution_by_crop["count"]
    / crop_totals
    * 100
).round(3)

distribution_export = pd.concat(
    [
        overall_distribution.assign(crop="ALL")[
            [
                "crop",
                "crop_stress_label",
                "count",
                "percentage",
            ]
        ],
        distribution_by_crop[
            [
                "crop",
                "crop_stress_label",
                "count",
                "percentage",
            ]
        ],
    ],
    ignore_index=True,
)

# ---------------------------------------------------------
# 7. Save outputs and provenance
# ---------------------------------------------------------

labeled["crop_stress_label"] = (
    labeled["crop_stress_label"].astype(str)
)

labeled.to_csv(
    output_path,
    index=False
)

distribution_export.to_csv(
    distribution_path,
    index=False
)

def sha256_file(path):
    digest = hashlib.sha256()

    with open(path, "rb") as file:
        for block in iter(
            lambda: file.read(1024 * 1024),
            b"",
        ):
            digest.update(block)

    return digest.hexdigest()

input_hash = sha256_file(input_path)
output_hash = sha256_file(output_path)

minimum_class_count = int(
    overall_distribution["count"].min()
)

report = f"""FarmSight Crop-Stress Target Definition Report
=================================================
Target type: operational yield-stress proxy

Baseline:
Median yield from the previous five observations
for the same state, district and crop.

Future observations used in baseline: No

Class definition:
High: yield anomaly <= -20 percent
Moderate: -20 percent < anomaly <= -10 percent
Low: anomaly > -10 percent

Input rows: {len(data)}
Yield-inconsistent rows excluded: {excluded_inconsistent}
Consistent rows: {len(consistent_data)}
Rows without five-history baseline excluded: {excluded_no_baseline}
Final labeled rows: {len(labeled)}

Duplicate district-year-crop keys: {duplicate_keys}
Missing target values: {target_missing}
Missing baseline values: {baseline_missing}
Infinite anomaly values: {infinite_anomalies}
Minimum class count: {minimum_class_count}

Important leakage restriction:
area_1000_ha, production_1000_tons,
yield_kg_per_ha, yield_baseline_previous_5 and
yield_anomaly_pct are retained only for auditing and
target construction. They must not be model inputs.

Input SHA-256:
{input_hash}

Output SHA-256:
{output_hash}
"""

report_path.write_text(
    report,
    encoding="utf-8"
)

print("\n" + report)

print("\nOverall label distribution:")
print(
    overall_distribution.to_string(index=False)
)

print("\nDistribution by crop:")
print(
    distribution_by_crop.to_string(index=False)
)

print("\nSaved labeled dataset:", output_path)
print("Saved distribution:", distribution_path)
print("Saved target report:", report_path)

checks_passed = (
    len(labeled) > 0
    and duplicate_keys == 0
    and target_missing == 0
    and baseline_missing == 0
    and infinite_anomalies == 0
    and minimum_class_count >= 100
)

if checks_passed:
    print("CROP-STRESS TARGET DATASET VERIFIED")
else:
    print("CROP-STRESS TARGET DATASET NEEDS REVIEW")
```

```python
%pip install -q xgboost joblib

from pathlib import Path
import hashlib
import json
import platform

import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import sklearn
import xgboost

from sklearn.base import clone
from sklearn.compose import ColumnTransformer
from sklearn.dummy import DummyClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    f1_score,
    log_loss,
)
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.utils.class_weight import compute_sample_weight
from xgboost import XGBClassifier

# ---------------------------------------------------------
# 1. Paths
# ---------------------------------------------------------

base = Path("/content/drive/MyDrive/FarmSight_XAI")

input_path = (
    base / "data/processed/"
    "farmsight_modeling_dataset_split.csv"
)

validation_path = (
    base / "results/tables/"
    "model_validation_comparison.csv"
)

test_metrics_path = (
    base / "results/tables/"
    "selected_model_test_metrics.csv"
)

classification_path = (
    base / "results/tables/"
    "selected_model_test_classification_report.csv"
)

predictions_path = (
    base / "results/tables/"
    "selected_model_test_predictions.csv"
)

confusion_path = (
    base / "results/figures/"
    "selected_model_test_confusion_matrix.png"
)

model_path = (
    base / "models/"
    "farmsight_selected_pipeline.joblib"
)

metadata_path = (
    base / "models/"
    "farmsight_selected_pipeline_metadata.json"
)

report_path = (
    base / "logs/"
    "model_training_evaluation_report.txt"
)

for path in [
    validation_path,
    test_metrics_path,
    classification_path,
    predictions_path,
    confusion_path,
    model_path,
    metadata_path,
    report_path,
]:
    path.parent.mkdir(parents=True, exist_ok=True)

# ---------------------------------------------------------
# 2. Load locked dataset
# ---------------------------------------------------------

data = pd.read_csv(input_path)

numeric_features = [
    "year",
    "district_latitude",
    "district_longitude",
    "annual_temp_mean_c",
    "annual_temp_max_c",
    "annual_temp_min_c",
    "annual_temp_range_c",
    "annual_precip_mm",
    "monsoon_precip_mm",
    "monsoon_temp_mean_c",
    "monthly_precip_rate_std_mm_day",
]

categorical_features = [
    "crop",
    "state_name",
]

model_features = (
    numeric_features
    + categorical_features
)

target_column = "crop_stress_label"

class_names = [
    "Low",
    "Moderate",
    "High",
]

label_to_number = {
    label: number
    for number, label in enumerate(class_names)
}

number_to_label = {
    number: label
    for label, number in label_to_number.items()
}

data["target_number"] = (
    data[target_column]
    .map(label_to_number)
)

if data["target_number"].isna().any():
    raise ValueError("Unexpected target label detected.")

data["target_number"] = (
    data["target_number"].astype(int)
)

train = data[data["split"] == "train"].copy()
validation = data[
    data["split"] == "validation"
].copy()
test = data[data["split"] == "test"].copy()

X_train = train[model_features]
y_train = train["target_number"]

X_validation = validation[model_features]
y_validation = validation["target_number"]

X_test = test[model_features]
y_test = test["target_number"]

print("Train shape:", X_train.shape)
print("Validation shape:", X_validation.shape)
print("Test shape:", X_test.shape)

# ---------------------------------------------------------
# 3. Leakage check
# ---------------------------------------------------------

forbidden_features = {
    "area_1000_ha",
    "production_1000_tons",
    "yield_kg_per_ha",
    "yield_consistency_flag",
    "percentage_difference",
    "yield_baseline_previous_5",
    "yield_anomaly_pct",
    "crop_stress_label",
    "split",
}

leakage_features_used = sorted(
    set(model_features)
    .intersection(forbidden_features)
)

if leakage_features_used:
    raise ValueError(
        f"Leakage features included: {leakage_features_used}"
    )

print("Leakage features used:", leakage_features_used)

# ---------------------------------------------------------
# 4. Preprocessing
# ---------------------------------------------------------

def build_preprocessor():

    numeric_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median"),
        ),
        (
            "scaler",
            StandardScaler(),
        ),
    ])

    categorical_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            ),
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False,
            ),
        ),
    ])

    return ColumnTransformer([
        (
            "numeric",
            numeric_pipeline,
            numeric_features,
        ),
        (
            "categorical",
            categorical_pipeline,
            categorical_features,
        ),
    ])

# ---------------------------------------------------------
# 5. Candidate models
# ---------------------------------------------------------

models = {
    "Dummy Prior": DummyClassifier(
        strategy="prior"
    ),

    "Logistic Regression": LogisticRegression(
        max_iter=3000,
        class_weight="balanced",
        random_state=42,
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=300,
        min_samples_leaf=2,
        max_features="sqrt",
        class_weight="balanced_subsample",
        n_jobs=-1,
        random_state=42,
    ),

    "XGBoost": XGBClassifier(
        n_estimators=400,
        max_depth=5,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        objective="multi:softprob",
        num_class=3,
        eval_metric="mlogloss",
        tree_method="hist",
        n_jobs=-1,
        random_state=42,
    ),
}

# ---------------------------------------------------------
# 6. Metric helper
# ---------------------------------------------------------

def calculate_metrics(
    true_values,
    predictions,
    probabilities,
):
    return {
        "accuracy": accuracy_score(
            true_values,
            predictions,
        ),
        "balanced_accuracy":
            balanced_accuracy_score(
                true_values,
                predictions,
            ),
        "macro_f1": f1_score(
            true_values,
            predictions,
            average="macro",
        ),
        "weighted_f1": f1_score(
            true_values,
            predictions,
            average="weighted",
        ),
        "log_loss": log_loss(
            true_values,
            probabilities,
            labels=[0, 1, 2],
        ),
    }

# ---------------------------------------------------------
# 7. Train and compare using validation data only
# ---------------------------------------------------------

validation_results = []
trained_candidates = {}

for model_name, classifier in models.items():

    print("\nTraining:", model_name)

    pipeline = Pipeline([
        (
            "preprocessor",
            build_preprocessor(),
        ),
        (
            "classifier",
            clone(classifier),
        ),
    ])

    fit_arguments = {}

    if model_name == "XGBoost":
        train_weights = compute_sample_weight(
            class_weight="balanced",
            y=y_train,
        )

        fit_arguments[
            "classifier__sample_weight"
        ] = train_weights

    pipeline.fit(
        X_train,
        y_train,
        **fit_arguments,
    )

    predictions = pipeline.predict(
        X_validation
    )

    probabilities = pipeline.predict_proba(
        X_validation
    )

    metrics = calculate_metrics(
        y_validation,
        predictions,
        probabilities,
    )

    validation_results.append({
        "model": model_name,
        **metrics,
    })

    trained_candidates[model_name] = pipeline

    print(
        "Validation macro-F1:",
        round(metrics["macro_f1"], 4),
    )

validation_comparison = pd.DataFrame(
    validation_results
).sort_values(
    by=[
        "macro_f1",
        "balanced_accuracy",
    ],
    ascending=False,
).reset_index(drop=True)

validation_comparison.to_csv(
    validation_path,
    index=False,
)

# Select only among actual ML models
eligible_models = validation_comparison[
    validation_comparison["model"]
    != "Dummy Prior"
].copy()

selected_model_name = (
    eligible_models.iloc[0]["model"]
)

print("\nSelected model:", selected_model_name)

# ---------------------------------------------------------
# 8. Refit selected model using train + validation
# ---------------------------------------------------------

development = pd.concat(
    [train, validation],
    ignore_index=True,
)

X_development = development[model_features]
y_development = development["target_number"]

selected_pipeline = Pipeline([
    (
        "preprocessor",
        build_preprocessor(),
    ),
    (
        "classifier",
        clone(models[selected_model_name]),
    ),
])

final_fit_arguments = {}

if selected_model_name == "XGBoost":
    development_weights = compute_sample_weight(
        class_weight="balanced",
        y=y_development,
    )

    final_fit_arguments[
        "classifier__sample_weight"
    ] = development_weights

selected_pipeline.fit(
    X_development,
    y_development,
    **final_fit_arguments,
)

# ---------------------------------------------------------
# 9. One-time untouched test evaluation
# ---------------------------------------------------------

test_predictions = selected_pipeline.predict(
    X_test
)

test_probabilities = (
    selected_pipeline.predict_proba(X_test)
)

test_metrics = calculate_metrics(
    y_test,
    test_predictions,
    test_probabilities,
)

test_metrics_table = pd.DataFrame([{
    "selected_model": selected_model_name,
    "test_rows": len(test),
    **test_metrics,
}])

test_metrics_table.to_csv(
    test_metrics_path,
    index=False,
)

class_report = classification_report(
    y_test,
    test_predictions,
    labels=[0, 1, 2],
    target_names=class_names,
    output_dict=True,
    zero_division=0,
)

class_report_table = (
    pd.DataFrame(class_report)
    .transpose()
    .reset_index()
    .rename(columns={"index": "class"})
)

class_report_table.to_csv(
    classification_path,
    index=False,
)

# ---------------------------------------------------------
# 10. Save test predictions
# ---------------------------------------------------------

prediction_table = test[
    [
        "state_code",
        "state_name",
        "district_code",
        "district_name",
        "year",
        "crop",
        target_column,
    ]
].copy()

prediction_table[
    "predicted_stress_label"
] = [
    number_to_label[int(value)]
    for value in test_predictions
]

prediction_table["probability_low"] = (
    test_probabilities[:, 0]
)

prediction_table["probability_moderate"] = (
    test_probabilities[:, 1]
)

prediction_table["probability_high"] = (
    test_probabilities[:, 2]
)

prediction_table.to_csv(
    predictions_path,
    index=False,
)

# ---------------------------------------------------------
# 11. Save confusion matrix
# ---------------------------------------------------------

matrix = confusion_matrix(
    y_test,
    test_predictions,
    labels=[0, 1, 2],
)

display = ConfusionMatrixDisplay(
    confusion_matrix=matrix,
    display_labels=class_names,
)

fig, axis = plt.subplots(
    figsize=(7, 6)
)

display.plot(
    ax=axis,
    cmap="Blues",
    colorbar=False,
)

axis.set_title(
    f"FarmSight Test Confusion Matrix\n"
    f"{selected_model_name}"
)

fig.tight_layout()

fig.savefig(
    confusion_path,
    dpi=300,
    bbox_inches="tight",
)

plt.show()
plt.close(fig)

# ---------------------------------------------------------
# 12. Save fitted model and metadata
# ---------------------------------------------------------

model_bundle = {
    "pipeline": selected_pipeline,
    "model_name": selected_model_name,
    "numeric_features": numeric_features,
    "categorical_features":
        categorical_features,
    "model_features": model_features,
    "class_names": class_names,
    "label_to_number": label_to_number,
    "number_to_label": number_to_label,
    "training_period":
        f"{int(development['year'].min())}-"
        f"{int(development['year'].max())}",
    "test_period":
        f"{int(test['year'].min())}-"
        f"{int(test['year'].max())}",
}

joblib.dump(
    model_bundle,
    model_path,
)

metadata = {
    "selected_model": selected_model_name,
    "selection_metric":
        "validation macro-F1",
    "random_seed": 42,
    "numeric_features": numeric_features,
    "categorical_features":
        categorical_features,
    "forbidden_features":
        sorted(forbidden_features),
    "class_names": class_names,
    "training_rows": len(train),
    "validation_rows": len(validation),
    "test_rows": len(test),
    "development_period": [
        int(development["year"].min()),
        int(development["year"].max()),
    ],
    "test_period": [
        int(test["year"].min()),
        int(test["year"].max()),
    ],
    "python_version": platform.python_version(),
    "pandas_version": pd.__version__,
    "scikit_learn_version":
        sklearn.__version__,
    "xgboost_version":
        xgboost.__version__,
}

metadata_path.write_text(
    json.dumps(
        metadata,
        indent=2,
    ),
    encoding="utf-8",
)

# ---------------------------------------------------------
# 13. Provenance report
# ---------------------------------------------------------

def sha256_file(path):
    digest = hashlib.sha256()

    with open(path, "rb") as file:
        for block in iter(
            lambda: file.read(1024 * 1024),
            b"",
        ):
            digest.update(block)

    return digest.hexdigest()

model_hash = sha256_file(model_path)
input_hash = sha256_file(input_path)

selected_validation_row = (
    validation_comparison[
        validation_comparison["model"]
        == selected_model_name
    ]
    .iloc[0]
)

report = f"""FarmSight Model Training and Evaluation Report
================================================
Selection criterion: validation macro-F1
Selected model: {selected_model_name}

Training rows: {len(train)}
Validation rows: {len(validation)}
Test rows: {len(test)}

Training period:
{int(train["year"].min())}-{int(train["year"].max())}

Validation period:
{int(validation["year"].min())}-{int(validation["year"].max())}

Test period:
{int(test["year"].min())}-{int(test["year"].max())}

Selected-model validation macro-F1:
{selected_validation_row["macro_f1"]:.6f}

Untouched-test metrics:
Accuracy: {test_metrics["accuracy"]:.6f}
Balanced accuracy: {test_metrics["balanced_accuracy"]:.6f}
Macro-F1: {test_metrics["macro_f1"]:.6f}
Weighted-F1: {test_metrics["weighted_f1"]:.6f}
Log loss: {test_metrics["log_loss"]:.6f}

Leakage features used: None
Random seed: 42

Input split dataset SHA-256:
{input_hash}

Saved model SHA-256:
{model_hash}
"""

report_path.write_text(
    report,
    encoding="utf-8",
)

# ---------------------------------------------------------
# 14. Final output
# ---------------------------------------------------------

print("\nVALIDATION MODEL COMPARISON")
print(
    validation_comparison.to_string(
        index=False
    )
)

print("\n" + report)

print("\nTEST CLASSIFICATION REPORT")
print(
    class_report_table.to_string(
        index=False
    )
)

required_outputs = [
    validation_path,
    test_metrics_path,
    classification_path,
    predictions_path,
    confusion_path,
    model_path,
    metadata_path,
    report_path,
]

all_outputs_exist = all(
    path.exists()
    and path.stat().st_size > 0
    for path in required_outputs
)

probability_sums_valid = bool(
    np.allclose(
        test_probabilities.sum(axis=1),
        1.0,
        atol=1e-6,
    )
)

print("\nAll artifacts exist:", all_outputs_exist)
print(
    "Probability rows sum to one:",
    probability_sums_valid,
)

if all_outputs_exist and probability_sums_valid:
    print(
        "MODEL TRAINING AND TEST EVALUATION VERIFIED"
    )
else:
    print(
        "MODEL EVALUATION NEEDS CORRECTION"
    )
```

```python
%pip install -q xgboost joblib

from pathlib import Path
import hashlib
import json

import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.compose import ColumnTransformer
from sklearn.dummy import DummyClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    f1_score,
)
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.utils.class_weight import compute_sample_weight
from xgboost import XGBClassifier

# =========================================================
# PATHS
# =========================================================

base = Path("/content/drive/MyDrive/FarmSight_XAI")

joined_path = (
    base / "data/processed/"
    "farmsight_crop_weather_1981_2017.csv"
)

weather_path = (
    base / "data/interim/"
    "nasa_power_district_annual_1981_2017.csv"
)

enhanced_data_path = (
    base / "data/processed/"
    "farmsight_enhanced_modeling_dataset.csv"
)

fold_metrics_path = (
    base / "results/tables/"
    "enhanced_model_fold_metrics.csv"
)

summary_path = (
    base / "results/tables/"
    "enhanced_model_cv_summary.csv"
)

class_report_path = (
    base / "results/tables/"
    "enhanced_selected_model_classification_report.csv"
)

confusion_path = (
    base / "results/figures/"
    "enhanced_model_cv_confusion_matrix.png"
)

model_path = (
    base / "models/"
    "farmsight_enhanced_selected_pipeline.joblib"
)

metadata_path = (
    base / "models/"
    "farmsight_enhanced_model_metadata.json"
)

report_path = (
    base / "logs/"
    "enhanced_model_experiment_report.txt"
)

for path in [
    enhanced_data_path,
    fold_metrics_path,
    summary_path,
    class_report_path,
    confusion_path,
    model_path,
    metadata_path,
    report_path,
]:
    path.parent.mkdir(parents=True, exist_ok=True)

# =========================================================
# PART 1 — CREATE PAST-ONLY FEATURES
# =========================================================

data = pd.read_csv(joined_path)
weather = pd.read_csv(weather_path)

consistency = (
    data["yield_consistency_flag"]
    .astype(str)
    .str.strip()
    .str.lower()
)

data = data[
    consistency.eq("true")
].copy()

group_keys = [
    "state_code",
    "district_code",
    "crop",
]

data = data.sort_values(
    group_keys + ["year"]
).reset_index(drop=True)

# Past yield information only
data["yield_lag_1"] = (
    data.groupby(group_keys)[
        "yield_kg_per_ha"
    ].shift(1)
)

data["past5_yield_mean"] = (
    data.groupby(group_keys)[
        "yield_kg_per_ha"
    ].transform(
        lambda values:
            values.shift(1)
            .rolling(5, min_periods=5)
            .mean()
    )
)

data["past5_yield_median"] = (
    data.groupby(group_keys)[
        "yield_kg_per_ha"
    ].transform(
        lambda values:
            values.shift(1)
            .rolling(5, min_periods=5)
            .median()
    )
)

data["past5_yield_std"] = (
    data.groupby(group_keys)[
        "yield_kg_per_ha"
    ].transform(
        lambda values:
            values.shift(1)
            .rolling(5, min_periods=5)
            .std(ddof=0)
    )
)

def rolling_slope(values):
    shifted = values.shift(1)

    return shifted.rolling(
        5,
        min_periods=5,
    ).apply(
        lambda window:
            np.polyfit(
                np.arange(5),
                window,
                1,
            )[0],
        raw=True,
    )

data["past5_yield_trend"] = (
    data.groupby(group_keys)[
        "yield_kg_per_ha"
    ].transform(rolling_slope)
)

# Target audit calculation
data["yield_anomaly_pct"] = (
    (
        data["yield_kg_per_ha"]
        - data["past5_yield_median"]
    )
    / data["past5_yield_median"]
    * 100
)

# Previous year's anomaly only
data["previous_yield_anomaly_pct"] = (
    data.groupby(group_keys)[
        "yield_anomaly_pct"
    ].shift(1)
)

data["crop_stress_label"] = np.select(
    [
        data["yield_anomaly_pct"] <= -20,
        (
            data["yield_anomaly_pct"] > -20
        )
        & (
            data["yield_anomaly_pct"] <= -10
        ),
    ],
    [
        "High",
        "Moderate",
    ],
    default="Low",
)

# =========================================================
# PART 2 — CREATE PAST-CLIMATE ANOMALIES
# =========================================================

weather_keys = [
    "state_code",
    "district_code",
    "year",
]

weather = weather.sort_values(
    weather_keys
).reset_index(drop=True)

climate_sources = [
    "annual_temp_mean_c",
    "annual_precip_mm",
    "monsoon_precip_mm",
]

for column in climate_sources:
    weather[f"past10_{column}"] = (
        weather
        .groupby(
            ["state_code", "district_code"]
        )[column]
        .transform(
            lambda values:
                values.shift(1)
                .rolling(
                    10,
                    min_periods=5,
                )
                .mean()
        )
    )

weather["annual_temp_anomaly_c"] = (
    weather["annual_temp_mean_c"]
    - weather["past10_annual_temp_mean_c"]
)

weather["annual_precip_anomaly_pct"] = (
    (
        weather["annual_precip_mm"]
        - weather["past10_annual_precip_mm"]
    )
    / weather["past10_annual_precip_mm"]
    * 100
)

weather["monsoon_precip_anomaly_pct"] = (
    (
        weather["monsoon_precip_mm"]
        - weather[
            "past10_monsoon_precip_mm"
        ]
    )
    / weather[
        "past10_monsoon_precip_mm"
    ]
    * 100
)

climate_features = weather[
    weather_keys + [
        "annual_temp_anomaly_c",
        "annual_precip_anomaly_pct",
        "monsoon_precip_anomaly_pct",
    ]
].copy()

data = data.merge(
    climate_features,
    on=weather_keys,
    how="left",
    validate="many_to_one",
)

# =========================================================
# PART 3 — LOCK ENHANCED FEATURES
# =========================================================

numeric_features = [
    "year",
    "district_latitude",
    "district_longitude",
    "annual_temp_mean_c",
    "annual_temp_max_c",
    "annual_temp_min_c",
    "annual_temp_range_c",
    "annual_precip_mm",
    "monsoon_precip_mm",
    "monsoon_temp_mean_c",
    "monthly_precip_rate_std_mm_day",
    "yield_lag_1",
    "past5_yield_mean",
    "past5_yield_median",
    "past5_yield_std",
    "past5_yield_trend",
    "previous_yield_anomaly_pct",
    "annual_temp_anomaly_c",
    "annual_precip_anomaly_pct",
    "monsoon_precip_anomaly_pct",
]

categorical_features = [
    "crop",
    "state_name",
]

model_features = (
    numeric_features
    + categorical_features
)

forbidden_features = {
    "yield_kg_per_ha",
    "production_1000_tons",
    "area_1000_ha",
    "yield_anomaly_pct",
    "crop_stress_label",
}

leakage_used = sorted(
    set(model_features)
    .intersection(forbidden_features)
)

if leakage_used:
    raise ValueError(
        f"Current-year leakage detected: "
        f"{leakage_used}"
    )

before_feature_filter = len(data)

data = data.replace(
    [np.inf, -np.inf],
    np.nan,
)

data = data.dropna(
    subset=(
        model_features
        + ["crop_stress_label"]
    )
).copy()

removed_for_missing_history = (
    before_feature_filter - len(data)
)

data = data.sort_values(
    ["year", "state_code", "district_code", "crop"]
).reset_index(drop=True)

duplicate_keys = int(
    data.duplicated(
        [
            "state_code",
            "district_code",
            "year",
            "crop",
        ]
    ).sum()
)

missing_features = int(
    data[model_features]
    .isna()
    .sum()
    .sum()
)

data.to_csv(
    enhanced_data_path,
    index=False,
)

print("Enhanced dataset rows:", len(data))
print(
    "Rows removed because past history was unavailable:",
    removed_for_missing_history,
)
print("Duplicate keys:", duplicate_keys)
print("Missing model features:", missing_features)
print("Current-year leakage features used:", leakage_used)

print("\nEnhanced target distribution:")
print(
    data["crop_stress_label"]
    .value_counts()
    .to_string()
)

# =========================================================
# PART 4 — EXPANDING-WINDOW VALIDATION
# =========================================================

class_names = [
    "Low",
    "Moderate",
    "High",
]

label_to_number = {
    label: number
    for number, label in enumerate(class_names)
}

data["target_number"] = (
    data["crop_stress_label"]
    .map(label_to_number)
    .astype(int)
)

folds = [
    {
        "name": "Fold 1",
        "train_end": 2001,
        "eval_start": 2002,
        "eval_end": 2005,
    },
    {
        "name": "Fold 2",
        "train_end": 2005,
        "eval_start": 2006,
        "eval_end": 2009,
    },
    {
        "name": "Fold 3",
        "train_end": 2009,
        "eval_start": 2010,
        "eval_end": 2013,
    },
    {
        "name": "Fold 4",
        "train_end": 2013,
        "eval_start": 2014,
        "eval_end": 2017,
    },
]

def build_preprocessor():

    numeric_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median"),
        ),
        (
            "scaler",
            StandardScaler(),
        ),
    ])

    categorical_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            ),
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False,
            ),
        ),
    ])

    return ColumnTransformer([
        (
            "numeric",
            numeric_pipeline,
            numeric_features,
        ),
        (
            "categorical",
            categorical_pipeline,
            categorical_features,
        ),
    ])

models = {
    "Dummy Prior": DummyClassifier(
        strategy="prior"
    ),

    "Logistic Regression": LogisticRegression(
        max_iter=3000,
        class_weight="balanced",
        random_state=42,
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=300,
        min_samples_leaf=2,
        max_features="sqrt",
        class_weight="balanced_subsample",
        n_jobs=-1,
        random_state=42,
    ),

    "XGBoost": XGBClassifier(
        n_estimators=400,
        max_depth=5,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        objective="multi:softprob",
        num_class=3,
        eval_metric="mlogloss",
        tree_method="hist",
        n_jobs=-1,
        random_state=42,
    ),
}

fold_results = []
oof_results = {
    model_name: {
        "true": [],
        "predicted": [],
    }
    for model_name in models
}

for fold in folds:

    train_fold = data[
        data["year"] <= fold["train_end"]
    ].copy()

    evaluation_fold = data[
        data["year"].between(
            fold["eval_start"],
            fold["eval_end"],
        )
    ].copy()

    if (
        len(train_fold) == 0
        or len(evaluation_fold) == 0
    ):
        raise ValueError(
            f"{fold['name']} is empty."
        )

    X_train = train_fold[model_features]
    y_train = train_fold["target_number"]

    X_evaluation = (
        evaluation_fold[model_features]
    )
    y_evaluation = (
        evaluation_fold["target_number"]
    )

    print(
        f"\n{fold['name']}: "
        f"train={len(train_fold)}, "
        f"evaluate={len(evaluation_fold)}"
    )

    for model_name, classifier in models.items():

        pipeline = Pipeline([
            (
                "preprocessor",
                build_preprocessor(),
            ),
            (
                "classifier",
                clone(classifier),
            ),
        ])

        fit_arguments = {}

        if model_name == "XGBoost":
            weights = compute_sample_weight(
                class_weight="balanced",
                y=y_train,
            )

            fit_arguments[
                "classifier__sample_weight"
            ] = weights

        pipeline.fit(
            X_train,
            y_train,
            **fit_arguments,
        )

        predictions = pipeline.predict(
            X_evaluation
        )

        accuracy = accuracy_score(
            y_evaluation,
            predictions,
        )

        balanced_accuracy = (
            balanced_accuracy_score(
                y_evaluation,
                predictions,
            )
        )

        macro_f1 = f1_score(
            y_evaluation,
            predictions,
            average="macro",
        )

        weighted_f1 = f1_score(
            y_evaluation,
            predictions,
            average="weighted",
        )

        fold_results.append({
            "fold": fold["name"],
            "train_end": fold["train_end"],
            "evaluation_period":
                f"{fold['eval_start']}-"
                f"{fold['eval_end']}",
            "model": model_name,
            "training_rows":
                len(train_fold),
            "evaluation_rows":
                len(evaluation_fold),
            "accuracy": accuracy,
            "balanced_accuracy":
                balanced_accuracy,
            "macro_f1": macro_f1,
            "weighted_f1": weighted_f1,
        })

        oof_results[model_name][
            "true"
        ].append(y_evaluation.to_numpy())

        oof_results[model_name][
            "predicted"
        ].append(predictions)

        print(
            model_name,
            "| accuracy:",
            round(accuracy, 4),
            "| macro-F1:",
            round(macro_f1, 4),
        )

fold_metrics = pd.DataFrame(
    fold_results
)

fold_metrics.to_csv(
    fold_metrics_path,
    index=False,
)

summary = (
    fold_metrics
    .groupby("model")
    .agg(
        folds=("fold", "count"),
        accuracy_mean=("accuracy", "mean"),
        accuracy_std=("accuracy", "std"),
        balanced_accuracy_mean=(
            "balanced_accuracy",
            "mean",
        ),
        macro_f1_mean=("macro_f1", "mean"),
        macro_f1_std=("macro_f1", "std"),
        weighted_f1_mean=(
            "weighted_f1",
            "mean",
        ),
    )
    .reset_index()
    .sort_values(
        [
            "macro_f1_mean",
            "balanced_accuracy_mean",
        ],
        ascending=False,
    )
)

summary.to_csv(
    summary_path,
    index=False,
)

eligible = summary[
    summary["model"] != "Dummy Prior"
].copy()

selected_model_name = (
    eligible.iloc[0]["model"]
)

# =========================================================
# PART 5 — POOLED CLASS REPORT
# =========================================================

selected_true = np.concatenate(
    oof_results[selected_model_name]["true"]
)

selected_predictions = np.concatenate(
    oof_results[selected_model_name][
        "predicted"
    ]
)

class_report = classification_report(
    selected_true,
    selected_predictions,
    labels=[0, 1, 2],
    target_names=class_names,
    output_dict=True,
    zero_division=0,
)

class_report_table = (
    pd.DataFrame(class_report)
    .transpose()
    .reset_index()
    .rename(columns={"index": "class"})
)

class_report_table.to_csv(
    class_report_path,
    index=False,
)

matrix = confusion_matrix(
    selected_true,
    selected_predictions,
    labels=[0, 1, 2],
)

display = ConfusionMatrixDisplay(
    confusion_matrix=matrix,
    display_labels=class_names,
)

fig, axis = plt.subplots(figsize=(7, 6))

display.plot(
    ax=axis,
    cmap="Greens",
    colorbar=False,
)

axis.set_title(
    "Enhanced Model Expanding-Window "
    "Confusion Matrix"
)

fig.tight_layout()
fig.savefig(
    confusion_path,
    dpi=300,
    bbox_inches="tight",
)

plt.show()
plt.close(fig)

# =========================================================
# PART 6 — TRAIN FINAL ENHANCED MODEL
# =========================================================

final_pipeline = Pipeline([
    (
        "preprocessor",
        build_preprocessor(),
    ),
    (
        "classifier",
        clone(models[selected_model_name]),
    ),
])

final_fit_arguments = {}

if selected_model_name == "XGBoost":
    final_weights = compute_sample_weight(
        class_weight="balanced",
        y=data["target_number"],
    )

    final_fit_arguments[
        "classifier__sample_weight"
    ] = final_weights

final_pipeline.fit(
    data[model_features],
    data["target_number"],
    **final_fit_arguments,
)

model_bundle = {
    "pipeline": final_pipeline,
    "model_name": selected_model_name,
    "experiment_type":
        "exploratory enhanced historical-feature model",
    "numeric_features": numeric_features,
    "categorical_features":
        categorical_features,
    "model_features": model_features,
    "class_names": class_names,
    "label_to_number": label_to_number,
    "number_to_label": {
        number: label
        for label, number
        in label_to_number.items()
    },
}

joblib.dump(
    model_bundle,
    model_path,
)

# =========================================================
# PART 7 — REPORT AND VERIFICATION
# =========================================================

selected_summary = summary[
    summary["model"] == selected_model_name
].iloc[0]

def sha256_file(path):
    digest = hashlib.sha256()

    with open(path, "rb") as file:
        for block in iter(
            lambda: file.read(1024 * 1024),
            b"",
        ):
            digest.update(block)

    return digest.hexdigest()

metadata = {
    "selected_model": selected_model_name,
    "selection_metric":
        "mean expanding-window macro-F1",
    "rows": len(data),
    "year_min": int(data["year"].min()),
    "year_max": int(data["year"].max()),
    "folds": folds,
    "model_features": model_features,
    "current_year_yield_used": False,
    "model_sha256": sha256_file(model_path),
}

metadata_path.write_text(
    json.dumps(metadata, indent=2),
    encoding="utf-8",
)

report = f"""FarmSight Enhanced Model Experiment
===================================
Experiment type:
Exploratory past-history feature experiment

Final enhanced rows: {len(data)}
Year range:
{int(data["year"].min())}-{int(data["year"].max())}

Current-year yield used as input: No
Current-year production used as input: No
Current-year cultivated area used as input: No

Past-only features included:
Previous yield, previous five-year yield statistics,
previous yield anomaly, historical yield trend and
climate anomalies.

Evaluation:
Four expanding-window temporal folds

Selected model: {selected_model_name}
Mean accuracy:
{selected_summary["accuracy_mean"]:.6f}

Mean balanced accuracy:
{selected_summary["balanced_accuracy_mean"]:.6f}

Mean macro-F1:
{selected_summary["macro_f1_mean"]:.6f}

Mean weighted-F1:
{selected_summary["weighted_f1_mean"]:.6f}

Important:
These are expanding-window cross-validation results.
They are not described as a new untouched test result.
"""

report_path.write_text(
    report,
    encoding="utf-8",
)

print("\nENHANCED MODEL SUMMARY")
print(summary.to_string(index=False))

print("\nSelected enhanced model:")
print(selected_model_name)

print("\nPOOLED CLASSIFICATION REPORT")
print(
    class_report_table.to_string(index=False)
)

print("\n" + report)

required_outputs = [
    enhanced_data_path,
    fold_metrics_path,
    summary_path,
    class_report_path,
    confusion_path,
    model_path,
    metadata_path,
    report_path,
]

all_outputs_exist = all(
    path.exists()
    and path.stat().st_size > 0
    for path in required_outputs
)

print("All enhanced artifacts exist:", all_outputs_exist)

if (
    all_outputs_exist
    and duplicate_keys == 0
    and missing_features == 0
    and not leakage_used
):
    print("ENHANCED MODEL EXPERIMENT VERIFIED")
else:
    print("ENHANCED MODEL EXPERIMENT NEEDS CORRECTION")
```

```python
%pip install -q shap lime

from pathlib import Path
import itertools
import json

import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import shap

from lime.lime_tabular import LimeTabularExplainer

# =========================================================
# PATHS
# =========================================================

base = Path("/content/drive/MyDrive/FarmSight_XAI")

data_path = (
    base / "data/processed/"
    "farmsight_enhanced_modeling_dataset.csv"
)

model_path = (
    base / "models/"
    "farmsight_enhanced_selected_pipeline.joblib"
)

shap_global_path = (
    base / "results/tables/"
    "enhanced_shap_global_importance.csv"
)

shap_class_path = (
    base / "results/tables/"
    "enhanced_shap_class_importance.csv"
)

shap_local_path = (
    base / "results/tables/"
    "enhanced_shap_local_explanation.csv"
)

lime_local_path = (
    base / "results/tables/"
    "enhanced_lime_local_explanation.csv"
)

quality_path = (
    base / "results/tables/"
    "enhanced_xai_quality_metrics.csv"
)

instance_path = (
    base / "results/tables/"
    "enhanced_xai_explained_instance.json"
)

shap_global_figure = (
    base / "results/figures/"
    "enhanced_shap_global_importance.png"
)

shap_local_figure = (
    base / "results/figures/"
    "enhanced_shap_local_explanation.png"
)

lime_figure = (
    base / "results/figures/"
    "enhanced_lime_local_explanation.png"
)

report_path = (
    base / "logs/"
    "enhanced_xai_analysis_report.txt"
)

outputs = [
    shap_global_path,
    shap_class_path,
    shap_local_path,
    lime_local_path,
    quality_path,
    instance_path,
    shap_global_figure,
    shap_local_figure,
    lime_figure,
    report_path,
]

for path in outputs:
    path.parent.mkdir(parents=True, exist_ok=True)

# =========================================================
# LOAD FINAL ENHANCED MODEL
# =========================================================

data = pd.read_csv(data_path)
bundle = joblib.load(model_path)

pipeline = bundle["pipeline"]
model_name = bundle["model_name"]
model_features = bundle["model_features"]
numeric_features = bundle["numeric_features"]
categorical_features = bundle["categorical_features"]
class_names = bundle["class_names"]
label_to_number = bundle["label_to_number"]

X = data[model_features].copy()

y = (
    data["crop_stress_label"]
    .map(label_to_number)
    .astype(int)
    .to_numpy()
)

predictions = pipeline.predict(X)
probabilities = pipeline.predict_proba(X)

preprocessor = pipeline.named_steps["preprocessor"]
classifier = pipeline.named_steps["classifier"]

transformed_feature_names = (
    preprocessor
    .get_feature_names_out()
    .tolist()
)

readable_transformed_names = [
    name
    .replace("numeric__", "")
    .replace("categorical__", "")
    for name in transformed_feature_names
]

print("Model explained:", model_name)
print("Enhanced rows:", len(data))
print(
    "Transformed features:",
    len(readable_transformed_names),
)

# =========================================================
# CHOOSE ONE CORRECT HIGH-STRESS EXAMPLE
# =========================================================

high_number = label_to_number["High"]

candidate_positions = np.where(
    (y == high_number)
    & (predictions == high_number)
)[0]

if len(candidate_positions) > 0:
    local_position = int(candidate_positions[0])
else:
    local_position = int(
        np.where(y == high_number)[0][0]
    )

local_row = X.iloc[[local_position]]
actual_number = int(y[local_position])
predicted_number = int(
    predictions[local_position]
)

actual_label = class_names[actual_number]
predicted_label = class_names[predicted_number]

# =========================================================
# SHAP — GLOBAL AND LOCAL
# =========================================================

sample_size = min(500, len(X))

sample_positions = (
    data
    .sample(
        n=sample_size,
        random_state=42,
    )
    .index
    .to_numpy()
)

X_shap = X.iloc[sample_positions]
X_shap_transformed = (
    preprocessor.transform(X_shap)
)

explainer = shap.TreeExplainer(classifier)

raw_shap = explainer.shap_values(
    X_shap_transformed
)

def normalize_shap(
    values,
    sample_count,
    feature_count,
    class_count,
):
    if isinstance(values, list):
        array = np.stack(values, axis=2)
    else:
        array = np.asarray(values)

    expected = (
        sample_count,
        feature_count,
        class_count,
    )

    if array.shape == expected:
        return array

    if array.shape == (
        class_count,
        sample_count,
        feature_count,
    ):
        return np.transpose(
            array,
            (1, 2, 0),
        )

    if array.shape == (
        sample_count,
        class_count,
        feature_count,
    ):
        return np.transpose(
            array,
            (0, 2, 1),
        )

    raise ValueError(
        f"Unsupported SHAP shape: {array.shape}"
    )

shap_values = normalize_shap(
    raw_shap,
    len(X_shap_transformed),
    len(readable_transformed_names),
    len(class_names),
)

print("Normalized SHAP shape:", shap_values.shape)

global_values = np.mean(
    np.abs(shap_values),
    axis=(0, 2),
)

global_table = pd.DataFrame({
    "feature": readable_transformed_names,
    "mean_absolute_shap": global_values,
}).sort_values(
    "mean_absolute_shap",
    ascending=False,
).reset_index(drop=True)

global_table["importance_percentage"] = (
    global_table["mean_absolute_shap"]
    / global_table["mean_absolute_shap"].sum()
    * 100
)

global_table.to_csv(
    shap_global_path,
    index=False,
)

class_records = []

for class_number, class_name in enumerate(
    class_names
):
    values = np.mean(
        np.abs(
            shap_values[:, :, class_number]
        ),
        axis=0,
    )

    for feature, importance in zip(
        readable_transformed_names,
        values,
    ):
        class_records.append({
            "class": class_name,
            "feature": feature,
            "mean_absolute_shap":
                float(importance),
        })

class_table = pd.DataFrame(
    class_records
).sort_values(
    ["class", "mean_absolute_shap"],
    ascending=[True, False],
)

class_table.to_csv(
    shap_class_path,
    index=False,
)

top_global = (
    global_table.head(15)
    .sort_values("mean_absolute_shap")
)

fig, axis = plt.subplots(figsize=(9, 7))

axis.barh(
    top_global["feature"],
    top_global["mean_absolute_shap"],
    color="#287a5c",
)

axis.set_xlabel("Mean absolute SHAP value")
axis.set_title(
    "Enhanced FarmSight Global SHAP Importance"
)

fig.tight_layout()
fig.savefig(
    shap_global_figure,
    dpi=300,
    bbox_inches="tight",
)

plt.show()
plt.close(fig)

# Local SHAP
local_transformed = (
    preprocessor.transform(local_row)
)

local_raw = explainer.shap_values(
    local_transformed
)

local_shap = normalize_shap(
    local_raw,
    1,
    len(readable_transformed_names),
    len(class_names),
)

local_contributions = (
    local_shap[
        0,
        :,
        predicted_number,
    ]
)

local_shap_table = pd.DataFrame({
    "feature": readable_transformed_names,
    "shap_contribution":
        local_contributions,
    "absolute_contribution":
        np.abs(local_contributions),
}).sort_values(
    "absolute_contribution",
    ascending=False,
).reset_index(drop=True)

local_shap_table.to_csv(
    shap_local_path,
    index=False,
)

top_local_shap = (
    local_shap_table.head(10)
    .sort_values("shap_contribution")
)

colors = [
    "#bd3d3a" if value < 0
    else "#287a5c"
    for value in top_local_shap[
        "shap_contribution"
    ]
]

fig, axis = plt.subplots(figsize=(9, 6))

axis.barh(
    top_local_shap["feature"],
    top_local_shap[
        "shap_contribution"
    ],
    color=colors,
)

axis.axvline(
    0,
    color="black",
    linewidth=0.8,
)

axis.set_xlabel(
    f"Contribution toward {predicted_label}"
)

axis.set_title(
    "Enhanced FarmSight Local SHAP Explanation"
)

fig.tight_layout()
fig.savefig(
    shap_local_figure,
    dpi=300,
    bbox_inches="tight",
)

plt.show()
plt.close(fig)

# =========================================================
# SHAP STABILITY
# =========================================================

rng = np.random.default_rng(42)
shap_top_sets = []

for _ in range(5):
    bootstrap = rng.choice(
        len(shap_values),
        size=len(shap_values),
        replace=True,
    )

    importance = np.mean(
        np.abs(shap_values[bootstrap]),
        axis=(0, 2),
    )

    top_indices = np.argsort(
        importance
    )[-10:]

    shap_top_sets.append(
        {
            readable_transformed_names[index]
            for index in top_indices
        }
    )

def mean_pairwise_jaccard(sets):
    scores = []

    for first, second in itertools.combinations(
        sets,
        2,
    ):
        union = len(first | second)

        scores.append(
            len(first & second) / union
            if union else 1.0
        )

    return float(np.mean(scores))

shap_stability = mean_pairwise_jaccard(
    shap_top_sets
)

shap_top5_share = float(
    global_table
    .head(5)["importance_percentage"]
    .sum()
)

# =========================================================
# LIME — ORIGINAL FEATURE SPACE
# =========================================================

lime_data = X.copy()

category_values = {}
category_indices = []
category_names = {}

for feature in categorical_features:
    values = sorted(
        lime_data[feature]
        .astype(str)
        .unique()
        .tolist()
    )

    category_values[feature] = values

    mapping = {
        value: index
        for index, value in enumerate(values)
    }

    lime_data[feature] = (
        lime_data[feature]
        .astype(str)
        .map(mapping)
        .astype(int)
    )

    feature_index = model_features.index(feature)

    category_indices.append(feature_index)
    category_names[feature_index] = values

lime_array = lime_data[
    model_features
].to_numpy(dtype=float)

background_size = min(
    3000,
    len(lime_array),
)

background_indices = (
    np.random.default_rng(42)
    .choice(
        len(lime_array),
        size=background_size,
        replace=False,
    )
)

lime_background = lime_array[
    background_indices
]

def lime_predict(array):
    frame = pd.DataFrame(
        array,
        columns=model_features,
    )

    for feature in categorical_features:
        values = category_values[feature]

        codes = (
            np.rint(frame[feature])
            .astype(int)
            .clip(0, len(values) - 1)
        )

        frame[feature] = [
            values[code]
            for code in codes
        ]

    for feature in numeric_features:
        frame[feature] = pd.to_numeric(
            frame[feature],
            errors="coerce",
        )

    return pipeline.predict_proba(
        frame[model_features]
    )

lime_explainer = LimeTabularExplainer(
    training_data=lime_background,
    feature_names=model_features,
    class_names=class_names,
    categorical_features=category_indices,
    categorical_names=category_names,
    mode="classification",
    discretize_continuous=True,
    random_state=42,
)

local_lime_row = lime_array[
    local_position
]

lime_explanation = (
    lime_explainer.explain_instance(
        data_row=local_lime_row,
        predict_fn=lime_predict,
        labels=[predicted_number],
        num_features=10,
        num_samples=3000,
    )
)

lime_items = lime_explanation.as_list(
    label=predicted_number
)

lime_table = pd.DataFrame(
    lime_items,
    columns=[
        "feature_condition",
        "lime_weight",
    ],
)

lime_table["absolute_weight"] = (
    lime_table["lime_weight"].abs()
)

lime_table.to_csv(
    lime_local_path,
    index=False,
)

lime_plot_table = (
    lime_table
    .sort_values("lime_weight")
)

lime_colors = [
    "#bd3d3a" if value < 0
    else "#287a5c"
    for value in lime_plot_table[
        "lime_weight"
    ]
]

fig, axis = plt.subplots(figsize=(9, 6))

axis.barh(
    lime_plot_table[
        "feature_condition"
    ],
    lime_plot_table["lime_weight"],
    color=lime_colors,
)

axis.axvline(
    0,
    color="black",
    linewidth=0.8,
)

axis.set_xlabel(
    f"LIME weight toward {predicted_label}"
)

axis.set_title(
    "Enhanced FarmSight Local LIME Explanation"
)

fig.tight_layout()
fig.savefig(
    lime_figure,
    dpi=300,
    bbox_inches="tight",
)

plt.show()
plt.close(fig)

lime_fidelity = float(
    lime_explanation.score
)

# =========================================================
# LIME STABILITY
# =========================================================

lime_top_sets = []

for seed in [42, 43, 44, 45, 46]:

    stability_explainer = (
        LimeTabularExplainer(
            training_data=lime_background,
            feature_names=model_features,
            class_names=class_names,
            categorical_features=
                category_indices,
            categorical_names=
                category_names,
            mode="classification",
            discretize_continuous=True,
            random_state=seed,
        )
    )

    stability_explanation = (
        stability_explainer.explain_instance(
            data_row=local_lime_row,
            predict_fn=lime_predict,
            labels=[predicted_number],
            num_features=10,
            num_samples=2000,
        )
    )

    terms = {
        term
        for term, weight
        in stability_explanation.as_list(
            label=predicted_number
        )
    }

    lime_top_sets.append(terms)

lime_stability = mean_pairwise_jaccard(
    lime_top_sets
)

# =========================================================
# SAVE INSTANCE AND QUALITY METRICS
# =========================================================

explained_row = data.iloc[
    local_position
]

instance = {
    "state":
        str(explained_row["state_name"]),
    "district":
        str(explained_row["district_name"]),
    "crop":
        str(explained_row["crop"]),
    "year":
        int(explained_row["year"]),
    "actual_label":
        actual_label,
    "predicted_label":
        predicted_label,
    "probability_low":
        float(probabilities[
            local_position, 0
        ]),
    "probability_moderate":
        float(probabilities[
            local_position, 1
        ]),
    "probability_high":
        float(probabilities[
            local_position, 2
        ]),
}

for feature in model_features:
    value = explained_row[feature]

    if isinstance(
        value,
        (np.integer, np.floating),
    ):
        value = value.item()

    instance[feature] = value

instance_path.write_text(
    json.dumps(instance, indent=2),
    encoding="utf-8",
)

quality = pd.DataFrame([{
    "model": model_name,
    "shap_sample_size": sample_size,
    "shap_bootstrap_repetitions": 5,
    "shap_top10_stability_jaccard":
        shap_stability,
    "shap_top5_importance_share_pct":
        shap_top5_share,
    "lime_samples": 3000,
    "lime_local_fidelity_r2":
        lime_fidelity,
    "lime_repetitions": 5,
    "lime_top10_stability_jaccard":
        lime_stability,
    "local_features_displayed": 10,
}])

quality.to_csv(
    quality_path,
    index=False,
)

report = f"""FarmSight Enhanced XAI Report
================================
Model: {model_name}

SHAP:
Sample size: {sample_size}
Top-10 stability Jaccard: {shap_stability:.6f}
Top-five importance share: {shap_top5_share:.3f}%

LIME:
Local fidelity R-squared: {lime_fidelity:.6f}
Top-10 stability Jaccard: {lime_stability:.6f}

Local example:
State: {instance["state"]}
District: {instance["district"]}
Crop: {instance["crop"]}
Year: {instance["year"]}
Actual label: {actual_label}
Predicted label: {predicted_label}

Interpretation restriction:
SHAP and LIME explain model behaviour.
They do not demonstrate causal relationships.
"""

report_path.write_text(
    report,
    encoding="utf-8",
)

# =========================================================
# FINAL OUTPUT
# =========================================================

print("\nTOP 15 GLOBAL SHAP FEATURES")
print(
    global_table.head(15).to_string(
        index=False
    )
)

print("\nEXPLAINED INSTANCE")
print(json.dumps(instance, indent=2))

print("\nLIME EXPLANATION")
print(lime_table.to_string(index=False))

print("\nXAI QUALITY METRICS")
print(quality.to_string(index=False))

all_outputs_exist = all(
    path.exists()
    and path.stat().st_size > 0
    for path in outputs
)

print("\nAll XAI artifacts exist:", all_outputs_exist)

if (
    all_outputs_exist
    and np.isfinite(lime_fidelity)
    and np.isfinite(shap_stability)
    and np.isfinite(lime_stability)
):
    print("ENHANCED SHAP AND LIME ANALYSIS VERIFIED")
else:
    print("ENHANCED XAI ANALYSIS NEEDS CORRECTION")
```

```python
%pip install -q xgboost joblib

from pathlib import Path
import json
import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from sklearn.base import clone
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestRegressor
from sklearn.impute import SimpleImputer
from sklearn.linear_model import Ridge
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    f1_score,
    mean_absolute_error,
)
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.utils.class_weight import compute_sample_weight
from xgboost import XGBRegressor

# =========================================================
# PATHS
# =========================================================

base = Path("/content/drive/MyDrive/FarmSight_XAI")

data_path = (
    base / "data/processed/"
    "farmsight_enhanced_modeling_dataset.csv"
)

classifier_model_path = (
    base / "models/"
    "farmsight_enhanced_selected_pipeline.joblib"
)

fold_path = (
    base / "results/tables/"
    "ordinal_regression_fold_metrics.csv"
)

summary_path = (
    base / "results/tables/"
    "ordinal_regression_summary.csv"
)

class_report_path = (
    base / "results/tables/"
    "ordinal_regression_classification_report.csv"
)

confusion_path = (
    base / "results/figures/"
    "ordinal_regression_confusion_matrix.png"
)

model_path = (
    base / "models/"
    "farmsight_ordinal_regression_pipeline.joblib"
)

report_path = (
    base / "logs/"
    "ordinal_regression_experiment_report.txt"
)

for path in [
    fold_path,
    summary_path,
    class_report_path,
    confusion_path,
    model_path,
    report_path,
]:
    path.parent.mkdir(parents=True, exist_ok=True)

# =========================================================
# LOAD DATA AND FEATURE DEFINITIONS
# =========================================================

data = pd.read_csv(data_path)

classifier_bundle = joblib.load(
    classifier_model_path
)

numeric_features = (
    classifier_bundle["numeric_features"]
)

categorical_features = (
    classifier_bundle["categorical_features"]
)

model_features = (
    classifier_bundle["model_features"]
)

class_names = [
    "Low",
    "Moderate",
    "High",
]

label_to_number = {
    "Low": 0,
    "Moderate": 1,
    "High": 2,
}

# Continuous ordered target
data["log_yield_ratio_target"] = np.log(
    data["yield_kg_per_ha"]
    / data["past5_yield_median"]
)

data["target_number"] = (
    data["crop_stress_label"]
    .map(label_to_number)
    .astype(int)
)

if not np.isfinite(
    data["log_yield_ratio_target"]
).all():
    raise ValueError(
        "Non-finite regression target detected."
    )

forbidden_inputs = {
    "yield_kg_per_ha",
    "yield_anomaly_pct",
    "log_yield_ratio_target",
    "crop_stress_label",
    "production_1000_tons",
    "area_1000_ha",
}

leakage_used = sorted(
    set(model_features)
    .intersection(forbidden_inputs)
)

if leakage_used:
    raise ValueError(
        f"Leakage inputs detected: {leakage_used}"
    )

print("Dataset rows:", len(data))
print("Current-year leakage inputs:", leakage_used)

# =========================================================
# PREPROCESSING AND MODELS
# =========================================================

def build_preprocessor():

    numeric_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(strategy="median"),
        ),
        (
            "scaler",
            StandardScaler(),
        ),
    ])

    categorical_pipeline = Pipeline([
        (
            "imputer",
            SimpleImputer(
                strategy="most_frequent"
            ),
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False,
            ),
        ),
    ])

    return ColumnTransformer([
        (
            "numeric",
            numeric_pipeline,
            numeric_features,
        ),
        (
            "categorical",
            categorical_pipeline,
            categorical_features,
        ),
    ])

models = {
    "Ridge Regression": Ridge(
        alpha=1.0
    ),

    "Random Forest Regressor":
        RandomForestRegressor(
            n_estimators=300,
            min_samples_leaf=2,
            max_features=0.7,
            n_jobs=-1,
            random_state=42,
        ),

    "XGBoost Regressor":
        XGBRegressor(
            n_estimators=400,
            max_depth=5,
            learning_rate=0.05,
            subsample=0.8,
            colsample_bytree=0.8,
            objective="reg:squarederror",
            tree_method="hist",
            n_jobs=-1,
            random_state=42,
        ),
}

folds = [
    {
        "name": "Fold 1",
        "train_end": 2001,
        "eval_start": 2002,
        "eval_end": 2005,
    },
    {
        "name": "Fold 2",
        "train_end": 2005,
        "eval_start": 2006,
        "eval_end": 2009,
    },
    {
        "name": "Fold 3",
        "train_end": 2009,
        "eval_start": 2010,
        "eval_end": 2013,
    },
    {
        "name": "Fold 4",
        "train_end": 2013,
        "eval_start": 2014,
        "eval_end": 2017,
    },
]

high_threshold = np.log(0.80)
moderate_threshold = np.log(0.90)

def convert_to_stress(log_ratio):
    return np.where(
        log_ratio <= high_threshold,
        2,
        np.where(
            log_ratio <= moderate_threshold,
            1,
            0,
        ),
    ).astype(int)

# =========================================================
# EXPANDING-WINDOW EVALUATION
# =========================================================

records = []

oof = {
    model_name: {
        "true": [],
        "predicted": [],
    }
    for model_name in models
}

for fold in folds:

    training = data[
        data["year"] <= fold["train_end"]
    ].copy()

    evaluation = data[
        data["year"].between(
            fold["eval_start"],
            fold["eval_end"],
        )
    ].copy()

    X_train = training[model_features]

    y_train_regression = training[
        "log_yield_ratio_target"
    ]

    y_train_class = training[
        "target_number"
    ]

    X_evaluation = evaluation[
        model_features
    ]

    y_evaluation_regression = evaluation[
        "log_yield_ratio_target"
    ].to_numpy()

    y_evaluation_class = evaluation[
        "target_number"
    ].to_numpy()

    training_weights = compute_sample_weight(
        class_weight="balanced",
        y=y_train_class,
    )

    print(
        f"\n{fold['name']}: "
        f"train={len(training)}, "
        f"evaluate={len(evaluation)}"
    )

    for model_name, regressor in models.items():

        pipeline = Pipeline([
            (
                "preprocessor",
                build_preprocessor(),
            ),
            (
                "regressor",
                clone(regressor),
            ),
        ])

        pipeline.fit(
            X_train,
            y_train_regression,
            regressor__sample_weight=
                training_weights,
        )

        predicted_log_ratio = pipeline.predict(
            X_evaluation
        )

        predicted_classes = convert_to_stress(
            predicted_log_ratio
        )

        actual_anomaly = (
            np.exp(
                y_evaluation_regression
            ) - 1
        ) * 100

        predicted_anomaly = (
            np.exp(
                np.clip(
                    predicted_log_ratio,
                    -5,
                    5,
                )
            ) - 1
        ) * 100

        metrics = {
            "accuracy": accuracy_score(
                y_evaluation_class,
                predicted_classes,
            ),
            "balanced_accuracy":
                balanced_accuracy_score(
                    y_evaluation_class,
                    predicted_classes,
                ),
            "macro_f1": f1_score(
                y_evaluation_class,
                predicted_classes,
                average="macro",
            ),
            "weighted_f1": f1_score(
                y_evaluation_class,
                predicted_classes,
                average="weighted",
            ),
            "anomaly_mae_percentage_points":
                mean_absolute_error(
                    actual_anomaly,
                    predicted_anomaly,
                ),
        }

        records.append({
            "fold": fold["name"],
            "train_end": fold["train_end"],
            "evaluation_period":
                f"{fold['eval_start']}-"
                f"{fold['eval_end']}",
            "model": model_name,
            "training_rows": len(training),
            "evaluation_rows":
                len(evaluation),
            **metrics,
        })

        oof[model_name]["true"].append(
            y_evaluation_class
        )

        oof[model_name][
            "predicted"
        ].append(predicted_classes)

        print(
            model_name,
            "| accuracy:",
            round(metrics["accuracy"], 4),
            "| macro-F1:",
            round(metrics["macro_f1"], 4),
        )

fold_metrics = pd.DataFrame(records)

fold_metrics.to_csv(
    fold_path,
    index=False,
)

summary = (
    fold_metrics
    .groupby("model")
    .agg(
        folds=("fold", "count"),
        accuracy_mean=("accuracy", "mean"),
        balanced_accuracy_mean=(
            "balanced_accuracy",
            "mean",
        ),
        macro_f1_mean=(
            "macro_f1",
            "mean",
        ),
        macro_f1_std=(
            "macro_f1",
            "std",
        ),
        weighted_f1_mean=(
            "weighted_f1",
            "mean",
        ),
        anomaly_mae_mean=(
            "anomaly_mae_percentage_points",
            "mean",
        ),
    )
    .reset_index()
    .sort_values(
        [
            "macro_f1_mean",
            "balanced_accuracy_mean",
        ],
        ascending=False,
    )
)

summary.to_csv(
    summary_path,
    index=False,
)

selected_name = summary.iloc[0]["model"]

# =========================================================
# POOLED CLASSIFICATION REPORT
# =========================================================

selected_true = np.concatenate(
    oof[selected_name]["true"]
)

selected_predictions = np.concatenate(
    oof[selected_name]["predicted"]
)

class_report = classification_report(
    selected_true,
    selected_predictions,
    labels=[0, 1, 2],
    target_names=class_names,
    output_dict=True,
    zero_division=0,
)

class_report_table = (
    pd.DataFrame(class_report)
    .transpose()
    .reset_index()
    .rename(columns={"index": "class"})
)

class_report_table.to_csv(
    class_report_path,
    index=False,
)

matrix = confusion_matrix(
    selected_true,
    selected_predictions,
    labels=[0, 1, 2],
)

display = ConfusionMatrixDisplay(
    confusion_matrix=matrix,
    display_labels=class_names,
)

fig, axis = plt.subplots(figsize=(7, 6))

display.plot(
    ax=axis,
    cmap="Purples",
    colorbar=False,
)

axis.set_title(
    "Ordinal Regression Expanding-Window "
    "Confusion Matrix"
)

fig.tight_layout()
fig.savefig(
    confusion_path,
    dpi=300,
    bbox_inches="tight",
)

plt.show()
plt.close(fig)

# =========================================================
# TRAIN AND SAVE FINAL REGRESSION MODEL
# =========================================================

final_pipeline = Pipeline([
    (
        "preprocessor",
        build_preprocessor(),
    ),
    (
        "regressor",
        clone(models[selected_name]),
    ),
])

final_weights = compute_sample_weight(
    class_weight="balanced",
    y=data["target_number"],
)

final_pipeline.fit(
    data[model_features],
    data["log_yield_ratio_target"],
    regressor__sample_weight=final_weights,
)

bundle = {
    "pipeline": final_pipeline,
    "model_name": selected_name,
    "prediction_mode":
        "ordinal regression converted to stress classes",
    "numeric_features": numeric_features,
    "categorical_features":
        categorical_features,
    "model_features": model_features,
    "class_names": class_names,
    "high_threshold_log_ratio":
        float(high_threshold),
    "moderate_threshold_log_ratio":
        float(moderate_threshold),
}

joblib.dump(bundle, model_path)

# =========================================================
# REPORT
# =========================================================

selected_summary = summary[
    summary["model"] == selected_name
].iloc[0]

report = f"""FarmSight Ordinal Regression Experiment
========================================
Purpose:
Predict continuous yield ratio and convert it
to Low, Moderate or High crop stress.

Selected model: {selected_name}

Evaluation:
Four expanding-window chronological folds

Mean accuracy:
{selected_summary["accuracy_mean"]:.6f}

Mean balanced accuracy:
{selected_summary["balanced_accuracy_mean"]:.6f}

Mean macro-F1:
{selected_summary["macro_f1_mean"]:.6f}

Mean weighted-F1:
{selected_summary["weighted_f1_mean"]:.6f}

Mean yield-anomaly MAE:
{selected_summary["anomaly_mae_mean"]:.6f}
percentage points

Current-year yield used as model input: No
Current-year yield used as training target: Yes

Current-year production used as input: No
Current-year cultivated area used as input: No

This is the final planned model-improvement
experiment. No further score chasing should be done.
"""

report_path.write_text(
    report,
    encoding="utf-8",
)

print("\nORDINAL REGRESSION SUMMARY")
print(summary.to_string(index=False))

print("\nSelected model:", selected_name)

print("\nPOOLED CLASSIFICATION REPORT")
print(
    class_report_table.to_string(index=False)
)

print("\n" + report)

required_outputs = [
    fold_path,
    summary_path,
    class_report_path,
    confusion_path,
    model_path,
    report_path,
]

all_outputs_exist = all(
    path.exists()
    and path.stat().st_size > 0
    for path in required_outputs
)

print("All ordinal artifacts exist:", all_outputs_exist)

if all_outputs_exist and not leakage_used:
    print("ORDINAL REGRESSION EXPERIMENT VERIFIED")
else:
    print("ORDINAL REGRESSION NEEDS CORRECTION")
```
